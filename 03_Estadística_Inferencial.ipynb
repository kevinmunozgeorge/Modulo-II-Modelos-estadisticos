# 1. Estimadores

## 1.1 Definición

En estadística inferencial, generalmente se busca conocer características desconocidas de una población a partir de la información obtenida de una muestra.

Un **estimador** es una regla o función que permite calcular una estimación de un parámetro poblacional utilizando los datos contenidos en una muestra.

En otras palabras, un estimador es una fórmula que transforma las observaciones muestrales en un valor numérico utilizado para aproximar un parámetro desconocido de la población.

Por ejemplo, la **media muestral** es un posible estimador puntual de la media poblacional $\mu$.

La media muestral se define como:

$$
\bar{Y}=\frac{1}{n}\sum_{i=1}^{n}Y_i
$$

donde:

- $\bar{Y}$ representa la media muestral.
- $n$ es el tamaño de la muestra.
- $Y_i$ representa cada una de las observaciones obtenidas en la muestra.

Esta expresión representa una regla de cálculo: primero se suman todas las observaciones de la muestra y posteriormente se divide el resultado entre el número total de observaciones.

El objetivo de un estimador es utilizar la información disponible de una muestra para obtener una aproximación del valor real de un parámetro poblacional, como la media $\mu$, una proporción $p$ o una varianza $\sigma^2$.

## 1.2 Propiedades

Cuando proponemos un estimador:

$$
\hat{\theta}=T(X_1,\dots,X_n)
$$

para aproximar un parámetro poblacional $\theta$, buscamos que dicho estimador tenga ciertas propiedades que garanticen que las estimaciones obtenidas sean confiables y representen adecuadamente el comportamiento de la población.

Las principales propiedades de los estimadores son:

### 1.2.1 Insegadez

Un estimador es **insesgado** cuando su valor esperado coincide con el verdadero valor del parámetro que se desea estimar.

Formalmente:

$$
E(\hat{\theta})=\theta
$$

O bien, en otras palabras, Si $\hat{u}$ es un estimador puntual de un parámetro $u$, entonces $\hat{u}$ es un estimador insesgado si

$$
E(\hat{u})=u.
$$

Si $E(\hat{u})\neq u$, se dice que $\hat{u}$ está sesgado.

Esto significa que, si se obtuvieran muchas muestras diferentes y se calculara el estimador en cada una de ellas, el promedio de las estimaciones estaría centrado en el verdadero valor del parámetro.

En términos prácticos, un estimador insesgado no presenta una tendencia sistemática a sobreestimar o subestimar el parámetro.

### 1.2.2 Consistencia

Un estimador es **consistente** cuando se aproxima al verdadero valor del parámetro conforme aumenta el tamaño de la muestra.

Formalmente:

$$
\hat{\theta}\xrightarrow{p}\theta
\qquad \text{cuando } n\rightarrow\infty
$$

Esto indica que al disponer de más información, la estimación se vuelve cada vez más precisa.

En términos prácticos, la consistencia garantiza que una muestra grande permite obtener estimaciones cercanas al valor real de la población.

### 1.2.3 Eficiencia

La **eficiencia** compara la precisión de diferentes estimadores que estiman el mismo parámetro.

Entre varios estimadores insesgados, se considera más eficiente aquel que tiene menor varianza:

$$
Var(\hat{\theta}_1)<Var(\hat{\theta}_2)
$$

Esto significa que sus estimaciones presentan menor dispersión alrededor del verdadero valor del parámetro.

En términos prácticos, un estimador eficiente proporciona resultados más estables y con menor incertidumbre.

En conjunto, las tres propiedades anteriores permiten evaluar la calidad de un estimador. Un buen estimador idealmente debe ser **insesgado**, **consistente** y **eficiente**, ya que estas características aseguran que las estimaciones obtenidas sean cercanas, estables y confiables.

## 1.3 Propiedades adicionales de los estimadores

Además de las propiedades clásicas de los estimadores (**insesgadez, consistencia y eficiencia**), existen otras características importantes dentro de la estadística inferencial, como la **suficiencia** y la **completitud**.

Estas propiedades no son requisitos obligatorios que todo estimador debe cumplir junto con las anteriores, sino que representan criterios adicionales para evaluar la información que contiene un estimador y sus características teóricas.

### 1.3.1 Completitud

Un estadístico suficiente $T$ es **completo** si no existe ninguna función no trivial $g(T)$ tal que su esperanza sea cero para todo valor de $\theta$:

$$
\mathbb{E}_\theta[g(T)] = 0 \quad \forall \theta \quad \Longrightarrow \quad g(T)=0\ \text{c.s.}
$$

- La completitud es una propiedad técnica importante para demostrar que un estimador es óptimo.

### 1.3.2 Suficiencia

Un estadístico $T(X_1,\dots,X_n)$ es **suficiente** para $\theta$ si **resume toda la información de la muestra sobre ese parámetro**.

- Criterio de factorización (Neyman–Fisher):  
  $T$ es suficiente si la función de verosimilitud $L(\theta; x)$ puede factorizarse como
  $$
  L(\theta; x) = g(T(x),\theta)\,h(x),
  $$
  donde $h$ no depende de $\theta$.
- Ejemplo:  
  En una normal $N(\mu,\sigma^2)$ con $\sigma^2$ conocida, la media $\bar X$ es suficiente para $\mu$.

### 1.3.3 UMVUE (Estimador Insesgado de Varianza Mínima)

El concepto de **UMVUE** (*Uniformly Minimum Variance Unbiased Estimator*) hace referencia al mejor estimador insesgado posible dentro de una familia de estimadores.

Si un estadístico $T$ es **suficiente y completo** para un parámetro $\theta$, y existe un estimador insesgado $\tilde{\theta}$, entonces el estimador:

$$
\hat{\theta}=\mathbb{E}[\tilde{\theta}\mid T]
$$

es un **UMVUE**, es decir, un estimador insesgado cuya varianza es menor o igual que la de cualquier otro estimador insesgado del mismo parámetro.

Esto significa que, entre todos los estimadores que no presentan sesgo, el UMVUE proporciona la mayor precisión posible al tener la menor dispersión alrededor del verdadero valor del parámetro.

## 1.4 Resumen de propiedades

- **Insesgadez:** el estimador tiene como promedio el verdadero valor del parámetro.
- **Consistencia:** la estimación se aproxima al parámetro poblacional cuando aumenta el tamaño de la muestra.
- **Eficiencia:** mide la precisión del estimador mediante la reducción de su varianza.
- **Suficiencia y completitud:** indican que el estadístico conserva toda la información relevante del parámetro y permite obtener estimadores óptimos.
- **UMVUE:** representa el estimador insesgado con la menor varianza posible, por lo que es el estimador más eficiente dentro de los estimadores insesgados.

# 2. Métodos para construir estimadores

Como ya lo mencionamos antes, en estadística inferencial, uno de los principales objetivos consiste en obtener estimadores que permitan aproximar parámetros desconocidos de una población utilizando la información contenida en una muestra.

Existen diferentes métodos para construir estimadores. Cada uno utiliza un enfoque distinto para obtener una regla de cálculo a partir de los datos observados. Entre algunos de los métodos más importantes se encuentran:

## 2.1 Método de momentos

El **método de momentos** consiste en igualar los momentos poblacionales con los momentos muestrales con el objetivo de obtener estimaciones de los parámetros desconocidos.

La idea principal es utilizar características de la distribución, como la esperanza y otros momentos, para construir ecuaciones que permitan despejar los parámetros.

Si una distribución depende de un parámetro $\theta$, y su primer momento es:

$$
E(X)=g(\theta),
$$

se reemplaza por el momento muestral:

$$
\bar X=\frac{1}{n}\sum_{i=1}^{n}X_i,
$$

y se resuelve la ecuación:

$$
\bar X=g(\theta).
$$

Este método proporciona estimadores sencillos de calcular y constituye uno de los primeros métodos desarrollados para la estimación de parámetros.

## 2.2 Estimación por máxima verosimilitud (EMV)

La **estimación por máxima verosimilitud** busca encontrar el valor del parámetro que hace que los datos observados sean lo más probables posible.

Para una muestra:

$$
X_1,X_2,\ldots,X_n,
$$

con función de densidad o probabilidad:

$$
f(x;\theta),
$$

se define la función de verosimilitud como:

$$
L(\theta)=\prod_{i=1}^{n}f(X_i;\theta).
$$

El estimador de máxima verosimilitud se obtiene como el valor del parámetro que maximiza esta función:

$$
\hat{\theta}_{MV}
=
\arg\max_{\theta}L(\theta).
$$

En la práctica, suele utilizarse el logaritmo de la función de verosimilitud:

$$
\ell(\theta)=\ln(L(\theta)),
$$

ya que transforma el producto en una suma y facilita los cálculos.

La EMV es uno de los métodos más utilizados en estadística debido a sus buenas propiedades teóricas, especialmente cuando el tamaño de muestra es grande.

### Ejemplo 1.

Lanzamos una moneda 100 veces, sabiendo que se obtienen 55 soles. Encontrar el EMV para la probabilidad $p$ de obtener sol en un solo lanzamiento.

Sea
$X$: el número de soles que se obtienen al lanzar una moneda
se tiene que $X\sim \text{Bin}(100,p)$, entonces

$$ f(x;p) = \binom{100}{55}p^{55}(1-p)^{100-55}$$

```python
from scipy.stats import binom
import numpy as np

# Vamos a utilizar la libreria Sympy para matemática símbolica
import sympy

# importa la variable símbolica x
from sympy.abc import x
```

```python
# Vamos a definir a p como una variable símbolica positiva: 0<p<1
p = sympy.symbols('p', positive= True)

# Definimos la función de verosimilitus
# se puede omitir la constante
f = p**55 * (1-p)**45

# Ahora derivamos f con respecto a p
df = sympy.diff(f,p)

# Hay que igualar la derivada a cero y despejar a p
# df=0 para encontrar p
phat = sympy.solve(df,p)[0] # [1] es la segunda solucion y [0] es la primera

print("El estimador de máxima verosimilitud es", phat)
```

Del resultado anterior podemos observar que lo obtenido corresponde a la proporción de soles.

### Ejemplo 2.

Suponga que la vida útil de los focos de cierta marca, se modela mediante una distribución exponencial de parámetro $\theta$ (es desconocido). Probamos 5 focos y encontramos que tienen una vida útil de $2,3,1,3$ y $5$ años respectivamente. Calcualar el EMV (MLE) para $\theta$.

```python
import sympy
import numpy as np

# Datos de la muestra
r = np.array([2,3,1,3,5])

# Definimos las variables simbólicas
x, theta = sympy.symbols('x theta', positive=True)

# Función de densidad de la exponencial
f = theta * sympy.exp(-theta*x)

# Producto de las densidades (verosimilitud)
J = np.prod([f.subs(x, i) for i in r])

# Log-verosimilitud de manera expandida
logJ = sympy.expand_log(sympy.log(J))

# Derivada e igualamos a cero
theta_gorro = sympy.solve(sympy.diff(logJ, theta), theta)[0]

print("--- Desarrollo Simbólico con SymPy ---")
print(f"1. Función de Verosimilitud L(θ): {J}")
print(f"2. Log-Verosimilitud ln L(θ): {sympy.expand_log(sympy.log(J))}")
print( "3.El estimador de máxima verosimilitud es", theta_gorro)
print( "O en valores con decimales, su equivalente es:", float(theta_gorro))
```

### Ejemplo 3

El tiempo de retardo para pacientes con enfermedad coronaria potencial se define como la duración entre el comienzo de un síntoma agudo cardiovascular y la llegada a la unidad de emergencia.

Supongamos que el tiempo de retardo tiene una distribución exponencial de parámetro $\theta$.

Se tienen los siguientes registros de retardo (en minutos) en la clinica para los primeros 20 pacientes:

$$ 525, \ 719, \ 2880, \ 150, \ 30, \ 251, \ 45, \ 858, \ 15, \ 47, \ 90, \ 56, \ 68, \ 6, \ 189, \ 180, \ 60, \ 60, \ 294, \ 747 $$

- Encontrar el estimador por el método de momentos para la media
- Encontrar el valor del estimador con los datos.

Se tiene que
$$ X : \text{tiempo de retardo (en minutos)}$$
y que $X\sim\text{Exponencial}(\theta)$.

Recordemos que $\mathbb{E}[X]=1/\theta$ y y que $\text{Var}(X)= 1 /\theta^{2}$. Además el primer momento muestra es $\bar{X}$ y despejando a $\theta$ se tiene que
$$\hat{\theta} = 1/\bar{X}$$

```python
import sympy
import numpy as np
from sympy.abc import x

# 1. Definimos theta como variable simbólica positiva
theta = sympy.symbols('theta', positive=True)

# 2. Definimos la función de densidad de la Exponencial
f = theta * sympy.exp(-theta * x)

# 3. Datos del problema
datos_pacientes = np.array([525, 719, 2880, 150, 30, 251, 45, 858, 15, 47, 90, 56, 68, 6, 189, 180, 60, 60, 294, 747])

# 4. Construimos la función de verosimilitud (L) como el producto de las densidades
L = np.prod([f.subs(x, i) for i in datos_pacientes])

# 5. Calculamos la log-verosimilitud (logL) y la expandimos
logL = sympy.expand_log(sympy.log(L))

# 6. Derivamos con respecto a theta e igualamos a cero para resolver
derivada = sympy.diff(logL, theta)
theta_hat_pacientes = sympy.solve(derivada, theta)[0]

print(f"El estimador de máxima verosimilitud es: {float(theta_hat_pacientes):.5f} por minuto")
```

Se tiene que $\hat{\theta} = \frac{1}{\bar{X}} \approx 0.00275$ por minuto.

Por lo tanto, estimamos que el tiempo promedio de retardo entre el inicio de los sintomas cardiovasculares y la llegada a la unidad de emergencia es de aproximadamente 363.5 minutos.

## 2.3 Otros métodos de estimación

Además del método de momentos y la máxima verosimilitud, existen otros enfoques para construir estimadores, como:

- **Método de mínimos cuadrados:** busca minimizar la diferencia entre los valores observados y los valores estimados por un modelo.
- **Métodos bayesianos:** incorporan información previa sobre el parámetro mediante una distribución inicial y combinan dicha información con los datos observados.
- **Estimadores basados en funciones de pérdida:** seleccionan el estimador que minimiza una medida de error esperada.

Cada método tiene aplicaciones específicas y sus ventajas dependen del problema estadístico que se desea resolver.

# Distribuciones discretas

## Distribución Bernoulli

$$
X\sim \mathrm{Ber}(p)
$$

- Parámetro:

$$
0<p<1
$$

- Estimador de máxima verosimilitud:

$$
\hat p=\bar X
$$

- Esperanza del estimador:

$$
E[\hat p]=p
$$

- Varianza del estimador:

$$
\mathrm{Var}(\hat p)=\frac{p(1-p)}{n}
$$


---

## Distribución Binomial

$$
X\sim \mathrm{B}(m,p)
$$

considerando que $m$ es conocido.

- Parámetro:

$$
0<p<1
$$

- Estimador de máxima verosimilitud:

$$
\hat p=\frac{\bar X}{m}
$$

- Esperanza:

$$
E[\hat p]=p
$$

- Varianza:

$$
\mathrm{Var}(\hat p)=\frac{p(1-p)}{mn}
$$


---

## Distribución Poisson

$$
X\sim \mathrm{Pois}(\lambda)
$$

- Parámetro:

$$
\lambda>0
$$

- Estimador de máxima verosimilitud:

$$
\hat\lambda=\bar X
$$

- Esperanza:

$$
E[\hat\lambda]=\lambda
$$

- Varianza:

$$
\mathrm{Var}(\hat\lambda)=\frac{\lambda}{n}
$$

# Distribuciones continuas

## Distribución Exponencial

$$
X\sim \mathrm{Exp}(\lambda)
$$

- Parámetro:

$$
\lambda>0
$$

- Estimador de máxima verosimilitud:

$$
\hat\lambda=\frac{1}{\bar X}
$$

- Esperanza:

$$
E[\hat\lambda]=\frac{n}{n-1}\lambda,
\qquad n>1
$$

- Varianza:

$$
\mathrm{Var}(\hat\lambda)
=
\frac{n^2}{(n-1)^2(n-2)}\lambda^2,
\qquad n>2
$$


---

## Distribución Gamma

$$
X\sim \Gamma(\alpha,\beta)
$$

considerando la parametrización forma-tasa.

- Parámetros:

$$
\alpha>0,\qquad \beta>0
$$

- Estimadores de máxima verosimilitud:

Si $\alpha$ es conocida:

$$
\hat\beta=\frac{\alpha}{\bar X}
$$

Si $\beta$ es conocida, $\hat\alpha$ se obtiene resolviendo:

$$
\psi(\alpha)-\ln(\alpha)
=
\overline{\ln X}-\ln(\bar X)
$$

Algunas propiedades del estimador son:

$$
E[\hat\beta]
=
\frac{n\alpha}{n\alpha-1}\beta
$$

y

$$
\mathrm{Var}(\hat\beta)
=
\frac{n^2\alpha^2}
{(n\alpha-1)^2(n\alpha-2)}
\beta^2
$$


---

## Distribución Normal con varianza conocida

Cuando:

$$
X\sim N(\mu,\sigma^2)
$$

y $\sigma^2$ es conocida.

- Parámetro a estimar:

$$
\mu\in\mathbb{R}
$$

- Estimador de máxima verosimilitud:

$$
\hat\mu=\bar X
$$

- Esperanza:

$$
E[\hat\mu]=\mu
$$

- Varianza:

$$
\mathrm{Var}(\hat\mu)=\frac{\sigma^2}{n}
$$


---

## Distribución Normal con media conocida

Cuando $\mu$ es conocida, el parámetro de interés es:

$$
\sigma^2>0
$$

- Estimador de máxima verosimilitud:

$$
\hat\sigma^2=
\frac{1}{n}
\sum (X_i-\mu)^2
$$

- Esperanza:

$$
E[\hat\sigma^2]=\sigma^2
$$

- Varianza:

$$
\mathrm{Var}(\hat\sigma^2)
=
\frac{2\sigma^4}{n}
$$


---

## Distribución Normal con media y varianza desconocidas

Cuando ambos parámetros son desconocidos, los estimadores de máxima verosimilitud son:

- Para la media:

$$
\hat\mu=\bar X
$$

- Para la varianza:

$$
\hat\sigma^2=
\frac{1}{n}
\sum (X_i-\bar X)^2
$$

Sus principales propiedades son:

- Esperanzas:

$$
E[\hat\mu]=\mu
$$

$$
E[\hat\sigma^2]=\frac{n-1}{n}\sigma^2
$$

- Varianzas:

$$
\mathrm{Var}(\hat\mu)=\frac{\sigma^2}{n}
$$

$$
\mathrm{Var}(\hat\sigma^2)
=
\frac{2(n-1)\sigma^4}{n^2}
$$

# 4. Intervalos de confianza

Un **intervalo de confianza** es un rango de valores construido a partir de una muestra que permite estimar un parámetro desconocido de una población con cierto nivel de confianza.

La idea principal es que, en lugar de obtener un único valor como estimación puntual, se proporciona un intervalo que refleja la incertidumbre asociada al proceso de muestreo.

Un intervalo de confianza para un parámetro $\theta$ tiene la forma:

$$
P(L \leq \theta \leq U)=1-\alpha
$$

donde:

- $L$ representa el límite inferior del intervalo.
- $U$ representa el límite superior del intervalo.
- $1-\alpha$ es el nivel de confianza.
- $\alpha$ representa el nivel de significancia.

Por ejemplo, un nivel de confianza del $95\%$ significa que, si el procedimiento de construcción del intervalo se repitiera muchas veces, aproximadamente el $95\%$ de los intervalos obtenidos contendrían al verdadero valor del parámetro poblacional.

## 4.1 Intervalo de confianza para una media

Cuando se desea estimar la media poblacional $\mu$ a partir de una muestra, el intervalo depende de si la desviación estándar poblacional es conocida o desconocida.

Cuando la varianza poblacional es desconocida y la muestra proviene de una población aproximadamente normal, se utiliza la distribución **t de Student**.

El intervalo de confianza está dado por:

$$
\bar X
\pm
t_{\alpha/2,n-1}
\frac{s}{\sqrt n}
$$

donde:

- $\bar X$ es la media muestral.
- $s$ es la desviación estándar de la muestra.
- $n$ es el tamaño de muestra.
- $t_{\alpha/2,n-1}$ es el valor crítico de la distribución t de Student.

### Ejemplo

Se toma una muestra de $25$ estudiantes y se registra su calificación promedio. Los resultados obtenidos son:

$$
\bar X=82
$$

y

$$
s=10.
$$

Construir un intervalo de confianza del $95\%$ para la media poblacional.

```python
# ============================================
# Intervalo de confianza para una media
# utilizando distribución t de Student
# ============================================

import numpy as np
from scipy.stats import t

# Datos
media = 82
desv = 10
n = 25

# Nivel de confianza
confianza = 0.95
alpha = 1 - confianza

# Valor crítico t
t_critico = t.ppf(1-alpha/2, df=n-1)

# Error estándar
error = desv / np.sqrt(n)

# Intervalo
limite_inferior = media - t_critico * error
limite_superior = media + t_critico * error

print(f"Intervalo de confianza: ({limite_inferior:.2f}, {limite_superior:.2f})")
```

El intervalo obtenido representa un rango de posibles valores para la verdadera media poblacional.

Con un nivel de confianza del $95\%$, se estima que la media real de las calificaciones de todos los estudiantes se encuentra dentro del intervalo calculado.

**Por lo tanto, el intervalo obtenido proporciona una estimación de la media poblacional considerando la incertidumbre asociada a la muestra utilizada.**

## 4.2 Intervalo de confianza para una proporción

Cuando el parámetro de interés es una proporción poblacional $p$, se utiliza la distribución normal como aproximación cuando el tamaño de muestra es suficientemente grande.

El intervalo de confianza para una proporción es:

$$
\hat p
\pm
z_{\alpha/2}
\sqrt{
\frac{\hat p(1-\hat p)}{n}
}
$$

donde:

- $\hat p$ es la proporción observada en la muestra.
- $n$ es el tamaño de muestra.
- $z_{\alpha/2}$ es el valor crítico de la distribución normal estándar.

### Ejemplo

En una encuesta realizada a $500$ personas, $320$ indicaron que utilizan una aplicación determinada.

La proporción muestral es:

$$
\hat p=\frac{320}{500}=0.64
$$

Construir un intervalo de confianza del $95\%$ para la proporción poblacional.

```python
# ============================================
# Intervalo de confianza para una proporción
# ============================================

import numpy as np
from scipy.stats import norm

# Datos
x = 320
n = 500

# Proporción muestral
p_hat = x/n

# Nivel de confianza
confianza = 0.95
alpha = 1-confianza

# Valor crítico normal
z = norm.ppf(1-alpha/2)

# Error estándar
error = np.sqrt(p_hat*(1-p_hat)/n)

# Intervalo
limite_inferior = p_hat - z*error
limite_superior = p_hat + z*error

print(f"Intervalo de confianza: ({limite_inferior:.4f}, {limite_superior:.4f})")
```

El intervalo obtenido representa los valores posibles para la proporción real de usuarios de la aplicación en la población.

Con un nivel de confianza del $95\%$, se estima que la verdadera proporción poblacional se encuentra dentro del intervalo calculado.

**Por lo tanto, el intervalo permite estimar una característica de toda la población utilizando únicamente la información obtenida de una muestra.**

## 4.3 Definición general de intervalo de confianza

Consideremos dos estadísticos obtenidos a partir de una muestra:

$$
T_1=t_1(X_1,\ldots,X_n)
$$

y

$$
T_2=t_2(X_1,\ldots,X_n)
$$

que cumplen la siguiente condición:

$$
\mathbb{P}(T_1\leq \theta \leq T_2)=\alpha
$$

Entonces, el intervalo definido por $(T_1,T_2)$ recibe el nombre de **intervalo de confianza del $100\alpha\%$** para el parámetro desconocido $\theta$.

Este intervalo representa un rango de valores calculado a partir de la muestra que contiene al verdadero valor del parámetro con una probabilidad asociada al nivel de confianza establecido.

## 4.4 Error estándar (SE)

Mide **cuánto varía** una estadística de una muestra a otra.

- Si la varianza poblacional es conocida:
$$
SE = \frac{\sigma}{\sqrt{n}}
$$

- Si la varianza poblacional es desconocida (usamos $s$, la desviación muestral):
$$
SE = \frac{s}{\sqrt{n}}
$$

> Entre mayor $n$, menor $SE$.

### Ejercicio 1

En una encuesta de una Universidad, 225 estudiantes son seleccionados al azar de los que toman cálculo y se les pregunta si cálculo es su materia favorita. 100 estudiantes responden que el cálculo es su materia favorita. Proporcione un intervalo de confianza del 95 %  para la proporción de todos los estudiantes que toman cálculo y que la consideran su materia favorita.

**Solución:**

Identificación de Parámetros:

* Tamaño de la muestra ($n$): $225$ estudiantes seleccionados al azar.

* Casos de éxito registrados ($x$): $100$ estudiantes responden que es su materia favorita.

* Proporción muestral ($\hat{p}$):
  $$\hat{p} = \frac{x}{n} = \frac{100}{225} = \frac{4}{9} \approx 0.4444$$

* Proporción complementaria ($\hat{q}$):
  $$\hat{q} = 1 - \hat{p} = 1 - 0.4444 = 0.5556$$

* Nivel de confianza deseado ($1 - \alpha$): $0.95 \implies \alpha = 0.05$

* Nivel de significancia en las colas ($\alpha/2$): $0.025$

Se cumplen las condiciones de una muestra grande ya que $n\hat{p} = 100 \ge 5$ y $n\hat{q} = 125 \ge 5$, por lo que ello usamos la distribución normal estándar.

La fórmula para el intervalo de confianza de una proporción poblacional $p$ es:
$$IC_{95\%}(p) = \hat{p} \pm z_{\alpha/2} \cdot SE$$

Donde:

* *$SE$ (Error Estándar):* Mide la variabilidad esperada del estimador y se calcula como:
  $$SE = \sqrt{\frac{\hat{p}(1-\hat{p})}{n}}$$

* *$z_{\alpha/2}$ (Valor Crítico):* Es el valor correspondiente en la distribución normal estándar para una confianza del $95\%$, el cual es $z_{0.025} = 1.96$.


*Cálculos:*

* *Calcular el Error Estándar ($SE$)*
  $$SE = \sqrt{\frac{0.4444 \cdot 0.5556}{225}} = \sqrt{\frac{0.2469}{225}} = \sqrt{0.001097} \approx 0.0331$$

* *Margen de Error ($ME$)*
  $$ME = z_{\alpha/2} \cdot SE = 1.96 \cdot 0.0331 \approx 0.0649$$

* *Límites del Intervalo*
  * Límite Inferior: $\hat{p} - ME = 0.4444 - 0.0649 = 0.3795$
  * Límite Superior: $\hat{p} + ME = 0.4444 + 0.0649 = 0.5093$

El intervalo resultante es:
$$IC_{95\%}(p) = (\mathbf{0.3795}, \mathbf{0.5093})$$

*Conclusión:*
Con un nivel de confianza del $95\%$, se estima que la proporción real de todos los estudiantes de la universidad que toman cálculo y consideran que es su materia favorita se encuentra entre el *$37.95\%$* y el *$50.93\%$*.

```python
import numpy as np
import scipy.stats as st
from math import sqrt

# Datos
n = 225
x = 100
confianza = 0.95

# Proporción muestral
p_gorro = x/n

# Error estándar
error_est = np.sqrt((p_hat*(1-p_hat))/n)

# Intervalo de confianza para la proporción
intervalo = st.norm.interval(confianza, p_gorro, error_est)

print(f'Proporción muestral = {p_gorro:.4f}')
print(f'El intervalo de confianza del {confianza*100}% es: ({intervalo[0]:.4f}, {intervalo[1]:.4f})')
```

### Ejercicio 2

Los datos $1.2,\ 2.1,\ 2.3,\ 1.9,\ 3.0,\ 1.5,\ 2.8,\ 2.4,\ 1.7,\ 2.6,\ 2.0,\ 1.8,\ 3.1,\ 2.2,\ 2.5$ se toman a partir de una distribución normal $N(\mu,\sigma^2)$ con $\mu$ desconocida.

**a)** Encuentra un intervalo de confianza al nivel de $90\%$ para $\mu$, dado que $\sigma =2$.

**b)** Encuentra un intervalo de confianza al nivel de $90\%$ para $\mu$.

**c)** Encuentra un intervalo de confianza al nivel de $90\%$ para $\sigma^2$.

```python
import statistics as st
from math import sqrt
from scipy import stats

datos = [1.2,2.1,2.3,1.9,3.0,1.5,2.8,2.4,1.7,2.6,2.0,1.8,3.1,2.2,2.5]
#a) Sólución:
n = len(datos)
xbar = st.mean(datos)
sigma = 2      # desviación poblacional conocida
conf = 0.90
alpha = 1 - conf
z = stats.norm.ppf(1 - alpha/2)

SE = sigma / sqrt(n)
LI = xbar - z*SE
LS = xbar + z*SE

print("Ejercicio 2(a) — IC90% para μ con σ=2")
#print(f"Media muestral = {xbar:.4f}")
#print(f"z = {z:.4f}")
#print(f"SE = {SE:.4f}")
print(f"IC90%: ({LI:.4f}, {LS:.4f})")

#b) Sólución:
s = st.pstdev(datos) * sqrt(n/(n-1))  # desviación muestral (ddof=1)
df = n - 1
t = stats.t.ppf(1 - alpha/2, df)

SE = s / sqrt(n)
LI = xbar - t*SE
LS = xbar + t*SE

print("Ejercicio 2(b) — IC90% para μ con σ desconocida (t)")
#print(f"s = {s:.4f}")
#print(f"t = {t:.4f}")
#print(f"SE = {SE:.4f}")
print(f"IC90%: ({LI:.4f}, {LS:.4f})")

#c) Sólución:
S2 = s**2
chi2_inf = stats.chi2.ppf(alpha/2, df)
chi2_sup = stats.chi2.ppf(1 - alpha/2, df)

LI = df*S2/chi2_sup
LS = df*S2/chi2_inf

print("Ejercicio 2(c) — IC90% para σ²")
#print(f"S² = {S2:.4f}")
#print(f"χ²_inf = {chi2_inf:.4f}, χ²_sup = {chi2_sup:.4f}")
print(f"IC90%: ({LI:.4f}, {LS:.4f})")
```

# 5. Distribuciones muestrales

Dentro de la estadística inferencial, el interés no se centra únicamente en observar los valores individuales obtenidos en una muestra, sino también en analizar ciertas funciones construidas a partir de dichos datos, conocidas como **estadísticos**. Algunos ejemplos de estos estadísticos son la media muestral, la varianza muestral y la proporción.

Para realizar estimaciones y construir intervalos de confianza es necesario estudiar el comportamiento de estos estadísticos cuando se obtienen diferentes muestras de una misma población. La distribución que describe este comportamiento recibe el nombre de **distribución muestral**.

---

**Definición**

Supongamos una población con distribución:

$$
F_\theta
$$

donde $\theta$ representa uno o varios parámetros desconocidos.

Al seleccionar una muestra aleatoria de tamaño $n$ se tiene:

$$
X_1,X_2,\dots,X_n \overset{iid}{\sim} F_\theta
$$

A partir de esta muestra se puede construir un estadístico:

$$
T=T(X_1,\dots,X_n)
$$

El estadístico $T$ es una cantidad calculada únicamente con la información de la muestra.

La **distribución muestral de $T$** es la distribución de probabilidad que describe todos los valores que puede tomar dicho estadístico al repetir muchas veces el proceso de muestreo bajo las mismas condiciones.

---

**Importancia de las distribuciones muestrales**

El estudio de las distribuciones muestrales es fundamental porque permite:

- Analizar la **variabilidad de los estimadores** obtenidos a partir de una muestra.
- Determinar la precisión de un estimador mediante el **error estándar**:

$$
SE(T)=\sqrt{\mathrm{Var}(T)}
$$

- Construir **intervalos de confianza** para estimar parámetros poblacionales.
- Realizar **pruebas de hipótesis** mediante el comportamiento esperado de los estadísticos.

En resumen, las distribuciones muestrales permiten conectar la información obtenida de una muestra con las características desconocidas de una población.

## 5.1 Distribución Ji-Cuadrada ($\chi^2$)

La distribución **Ji-Cuadrada** surge a partir de la suma de los cuadrados de variables aleatorias normales estándar.

Si una variable aleatoria:

$$
X\sim N(0,1)
$$

entonces el cuadrado de dicha variable sigue una distribución Ji-Cuadrada con un grado de libertad:

$$
X^2\sim \chi^2(1)
$$

De manera general, si se tienen $n$ variables aleatorias independientes e idénticamente distribuidas con distribución normal estándar:

$$
X_1,X_2,\dots,X_n \overset{iid}{\sim} N(0,1)
$$

entonces la suma de sus cuadrados tiene una distribución Ji-Cuadrada con $n$ grados de libertad:

$$
\sum_{i=1}^{n}X_i^2\sim\chi^2(n)
$$

donde $n$ representa el número de grados de libertad de la distribución.

La distribución Ji-Cuadrada es ampliamente utilizada en estadística inferencial, especialmente en la construcción de intervalos de confianza para varianzas y en pruebas de hipótesis relacionadas con la dispersión de una población.

### 5.1.1 Visualización de la distribución Ji-Cuadrada

La distribución Ji-Cuadrada depende del número de **grados de libertad** ($df$), los cuales determinan la forma de la curva de densidad.

En el siguiente ejemplo se grafican diferentes distribuciones $\chi^2$ con distintos valores de grados de libertad. Esto permite observar cómo cambia la distribución conforme aumenta la cantidad de variables normales que intervienen en la suma de cuadrados.

Al incrementar los grados de libertad:

- La distribución deja de estar tan concentrada cerca de cero.
- La curva se desplaza hacia valores mayores.
- La distribución adquiere una forma más simétrica conforme aumenta $df$.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import chi2

# Definimos los valores del eje x para evaluar la densidad
x_chi = np.linspace(0, 20, 500)

# Configuración de la figura
plt.figure(figsize=(10, 5))

# Diferentes grados de libertad a comparar
grados_libertad = [1, 3, 6, 10]

# Graficamos la función de densidad para cada caso
for df in grados_libertad:
    densidad = chi2.pdf(x_chi, df)
    plt.plot(x_chi, densidad, label=f"$\\chi^2$ con df={df}")

# Personalización de la gráfica
plt.title("Funciones de densidad de la distribución Ji-Cuadrada")
plt.xlabel("Valores de x")
plt.ylabel("Densidad")

plt.legend()
plt.grid(True)
plt.show()
```

## 5.2 Distribución $t$ de Student

La distribución **$t$ de Student** aparece cuando se desea realizar inferencia sobre la media de una población normal y la varianza poblacional es desconocida. Su construcción surge al combinar una variable normal estándar con una variable Ji-Cuadrada independiente.

Sean las variables aleatorias:

$$
X\sim N(0,1)
$$

y

$$
Y\sim \chi^2(n-1)
$$

donde $Y$ representa una variable con distribución Ji-Cuadrada con $n-1$ grados de libertad.

Entonces, la variable definida como:

$$
T=\frac{X}{\sqrt{\frac{Y}{n-1}}}
$$

sigue una distribución $t$ de Student con $n-1$ grados de libertad:

$$
T\sim t(n-1)
$$

La distribución $t$ tiene una forma similar a la distribución normal estándar, pero presenta colas más pesadas debido a la incertidumbre adicional causada por la estimación de la varianza poblacional.

A medida que aumenta el número de grados de libertad, la distribución $t$ se aproxima cada vez más a la distribución normal estándar:

$$
\lim_{n\rightarrow\infty}t(n)=N(0,1)
$$

Por esta razón, la distribución $t$ es ampliamente utilizada para construir intervalos de confianza y realizar pruebas de hipótesis sobre medias cuando la muestra es pequeña o la varianza poblacional es desconocida.

### 5.2.1 Comparación entre la distribución $t$ de Student y la normal estándar

La distribución $t$ de Student depende del número de **grados de libertad** y presenta una mayor dispersión que la distribución normal estándar cuando la muestra es pequeña.

En la siguiente simulación se comparan diferentes distribuciones $t$ con distintos valores de grados de libertad junto con la distribución normal estándar:

- Para pocos grados de libertad, la distribución $t$ presenta colas más amplias, lo que refleja una mayor incertidumbre en la estimación de la varianza.
- Conforme aumenta el número de grados de libertad, la distribución $t$ se concentra más alrededor de cero.
- Cuando los grados de libertad son suficientemente grandes, la distribución $t$ se aproxima a la distribución normal estándar.

Esta propiedad explica por qué la distribución $t$ es utilizada principalmente en muestras pequeñas y por qué, en muestras grandes, sus resultados son similares a los obtenidos utilizando la distribución normal.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import t, norm

# Valores del eje horizontal
x = np.linspace(-5, 5, 500)

# Grados de libertad seleccionados
grados_libertad = [1, 3, 10, 30]

# Colores para distinguir cada curva
colores = ['blue', 'orange', 'green', 'red']

plt.figure(figsize=(10, 5))

# Dibujar distribuciones t de Student
for df, color in zip(grados_libertad, colores):
    plt.plot(
        x,
        t.pdf(x, df),
        label=f"$t$ (df={df})",
        color=color
    )

# Dibujar distribución normal estándar como referencia
plt.plot(
    x,
    norm.pdf(x),
    'k--',
    label="Normal estándar"
)

# Configuración de la gráfica
plt.title("Distribución $t$ de Student para diferentes grados de libertad")
plt.xlabel("x")
plt.ylabel("Densidad de probabilidad")

plt.legend()
plt.grid(True)
plt.show()
```

# 1. Variable aleatoria

En esta primera sección definiremos a una variable aleatoria como una función del espacio muestral en el conjunto de números reales. Esto nos permitirá considerar que el resultado del experimento aleatorio es el número real tomado
por la variable aleatoria. En consecuencia, nuestro interés en el estudio de
los experimentos aleatorios se trasladará al estudio de las distintas variables
aleatorias y sus características particulares.

Consideremos que tenemos un experimento aleatorio cualquiera, junto con un espacio de probabilidad asociado:

$$
(\Omega,\mathcal{F},P)
$$


Una **variable aleatoria** es una transformación $X$ del espacio de resultados $\Omega$ al conjunto de números reales, esto es:

$$
X:\Omega \rightarrow \mathbb{R},
$$

tal que para cualquier número real $x$,

$$
\{\omega \in \Omega : X(\omega)\leq x\}\in\mathcal{F}.
$$

### Medida de probabilidad inducida

Para cualquier intervalo de la forma:

$$
(-\infty,x]
$$

se puede obtener su imagen inversa bajo \(X\), es decir:

$$
X^{-1}((-\infty,x])=\{\omega\in\Omega:X(\omega)\leq x\}.
$$

Como este conjunto pertenece a $\mathcal{F}$ por la condición de la definición de variable aleatoria, se puede aplicar la medida de probabilidad $P$, pues ésta tiene como dominio a $\mathcal{F}$.

Así, mediante la función $X$, puede trasladarse la medida de probabilidad $P$ a intervalos de la forma:

$
(-\infty,x].
$

Además, puede demostrarse que esta construcción es suficiente para extender dicha medida a la totalidad de la $\sigma$-álgebra de Borel de $\mathbb{R}$.

Esta medida de probabilidad inducida por la variable aleatoria $X$ permite definir la distribución de probabilidad asociada a la variable aleatoria.

### Ejemplo

Sea $c$ una constante. Para cualquier experimento aleatorio con espacio muestral $\Omega$, se puede definir la función constante:

$
X(\omega)=c.
$

Esto significa que, sin importar cuál sea el resultado obtenido del experimento aleatorio, la variable aleatoria $X$ siempre asignará el mismo valor $c$.

Por lo tanto, todos los elementos del espacio muestral son transformados mediante $X$ al número $c$.

Decimos entonces que $X$ es una **variable aleatoria constante**.

Para cualquier conjunto de Borel $A$ de $\mathbb{R}$, se cumple que:

$$
P(X\in A)=
\begin{cases}
1, & \text{si } c\in A,\\
0, & \text{si } c\notin A.
\end{cases}
$$

Es decir, la probabilidad de que $X$ pertenezca a un conjunto $A$ es igual a 1 si el valor constante $c$ está dentro de dicho conjunto, y es igual a 0 en caso contrario.

Esta variable aleatoria representa el caso más sencillo posible, ya que no existe incertidumbre en el valor que toma $X$.

# 2. Variables aleatorias discretas y continuas

Considerando el conjunto de valores que una variable aleatoria puede tomar, vamos a clasificar a las variables aleatorias en dos tipos: **discretas o continuas**.

Decimos que una variable aleatoria es **discreta** cuando el conjunto de valores que ésta toma es un conjunto discreto, es decir, un conjunto finito o numerable.

Por ejemplo, el conjunto:

$$
\{0,1,2,\dots,n\}
$$

es un conjunto discreto porque es finito, lo mismo que:

$$
\mathbb{N}
$$

pues, aunque éste es un conjunto infinito, es numerable y por lo tanto discreto.

Por otra parte, decimos preliminarmente que una variable aleatoria es **continua** cuando toma todos los valores dentro de un intervalo:

$$
(a,b)\subseteq\mathbb{R}.
$$

## 3.1 Función de probabilidad

Consideremos primero el **caso discreto**. Debe observarse que los conceptos que se exponen a continuación no hacen referencia a ningún experimento aleatorio en particular.

Sea $X$ una variable aleatoria discreta que toma los valores:

$$
x_0,x_1,x_2,\dots
$$

con probabilidades:

$$
p_0=P(X=x_0)
$$

$$
p_1=P(X=x_1)
$$

$$
p_2=P(X=x_2)
$$

$$
\vdots
$$

Esta lista de valores numéricos y sus probabilidades puede ser finita o infinita, pero numerable.

La **función de probabilidad** de $X$ se define como aquella función que toma estas probabilidades como valores.

**Definición**

Sea $X$ una variable aleatoria discreta con valores:

$$
x_0,x_1,\dots
$$

La función de probabilidad de $X$, denotada por:

$$
f(x):\mathbb{R}\rightarrow\mathbb{R},
$$

se define como sigue:

$$
f(x)=
\begin{cases}
P(X=x), & \text{si } x=x_0,x_1,\dots\\
0, & \text{en otro caso}.
\end{cases}
$$

En otras palabras, la función de probabilidad es simplemente aquella función que indica la probabilidad asociada a los distintos valores que puede tomar la variable aleatoria.

## 3.2 Valor esperado de una variable aleatoria discreta (Esperanza)

Sea $Y$ una variable aleatoria discreta con función de probabilidad $p(y)$.

Entonces, el **valor esperado** de $Y$, denotado por $\mathbb{E}(Y)$, se define como:

$$
\mathbb{E}(Y)=\sum_y y\,p(y).
$$

El valor esperado representa el promedio ponderado de los posibles valores que puede tomar la variable aleatoria, donde cada valor es ponderado por su probabilidad de ocurrencia.

## 3.3 Varianza

Si $Y$ es una variable aleatoria con media:

$$
E(Y)=m,
$$

la **varianza** de una variable aleatoria $Y$ se define como el valor esperado de:

$$
(Y-m)^2.
$$

Esto es:

$$
V(Y)=E[(Y-m)^2].
$$

La varianza mide la dispersión de los valores de la variable aleatoria alrededor de su media.

La **desviación estándar** de $Y$ es la raíz cuadrada positiva de $V(Y)$:

$$
\sigma_Y=\sqrt{V(Y)}.
$$

La desviación estándar permite expresar la variabilidad de la variable aleatoria en las mismas unidades que sus valores originales.

## 3.4 Uniforme discreta

Modela un experimento con $n$ resultados igualmente probables, por ejemplo $X \in \{1, 2, \dots, n\}$.

$$p(x) = \frac{1}{n}, \quad x = 1, \dots, n \qquad\qquad E[X] = \frac{n+1}{2} \qquad\qquad \text{Var}(X) = \frac{n^2 - 1}{12}$$

Ejemplo: el resultado de lanzar un dado justo de 6 caras, $X \in \{1, \dots, 6\}$.

### Ejercicio 1

## 3.5 Distribución Bernoulli

Un ensayo Bernoulli se define como aquel experimento aleatorio con únicamente dos posibles resultados, llamados genéricamente: **éxito** y **fracaso**.

Supondremos que las probabilidades de estos resultados son $p$ y $1-p$, respectivamente.

Si se define la variable aleatoria $X$ como aquella función que lleva el resultado éxito al número $1$ y el resultado fracaso al número $0$, entonces decimos que $X$ tiene una distribución Bernoulli con parámetro $p\in(0,1)$ y escribimos:

$$
X\sim Ber(p)
$$

La función de probabilidad se puede escribir de la siguiente forma:

$$
f(x)=
\begin{cases}
1-p, & \text{si } x=0,\\
p, & \text{si } x=1,\\
0, & \text{en otro caso}.
\end{cases}
$$

O bien, de manera compacta:

$$
f(x)=
\begin{cases}
p^x(1-p)^{1-x}, & \text{si } x=0,1,\\
0, & \text{en otro caso}.
\end{cases}
$$

La función de probabilidad indica que la variable aleatoria únicamente puede tomar dos valores:

- $X=1$, cuando ocurre un éxito.
- $X=0$, cuando ocurre un fracaso.

La esperanza de una variable aleatoria Bernoulli está dada por:

$$
E(X)=p
$$

mientras que su varianza es:

$$
Var(X)=p(1-p)
$$

La función de distribución acumulada para una variable aleatoria Bernoulli tiene la siguiente forma:

$$
F(x)=
\begin{cases}
0, & \text{si } x<0,\\
1-p, & \text{si } 0\leq x<1,\\
1, & \text{si } x\geq1.
\end{cases}
$$

```python
p = 0.05
dist_bernoulli = stats.bernoulli(p)

x_vals = np.array([0, 1])
print("Funcion de probabilidad p(x):", dist_bernoulli.pmf(x_vals))
print("Esperanza teorica E[X] =", dist_bernoulli.mean(), " (formula: p =", p, ")")
print("Varianza teorica Var(X) =", round(dist_bernoulli.var(), 4), " (formula: p(1-p) =", round(p * (1 - p), 4), ")")
```

```python
plt.figure()
plt.bar(x_vals, dist_bernoulli.pmf(x_vals), color="tab:blue", tick_label=["No defectuoso (0)", "Defectuoso (1)"])
plt.ylabel("P(X = x)")
plt.title("Bernoulli: probabilidad de un articulo defectuoso (p = 0.05)")
plt.show()
```

## 3.6 Distribución Binomial

Supongamos que efectuamos una serie de $n$ ensayos independientes Bernoulli en donde la probabilidad de éxito en cada ensayo es $p$.

Si denotamos por $E$ el resultado éxito y por $F$ el resultado fracaso, entonces el espacio muestral de este experimento consiste de todas las posibles sucesiones de longitud $n$ de caracteres $E$ y $F$.

Así, el espacio muestral consiste de:

$$
2^n
$$

elementos.

Si ahora definimos la variable aleatoria $X$ como aquella función que indica el número de éxitos en cada una de estas sucesiones, esto es:

$$
X(EE\cdots EE)=n
$$

$$
X(FE\cdots EE)=n-1
$$

$$
\vdots
$$

$$
X(FF\cdots FF)=0
$$

entonces tenemos que $X$ puede tomar los valores:

$$
0,1,2,\dots,n
$$

con las probabilidades dadas por la función de probabilidad:

$$
f(x)=
\begin{cases}
\binom{n}{x}p^x(1-p)^{n-x}, & \text{si } x=0,1,\dots,n,\\
0, & \text{en otro caso}.
\end{cases}
$$

Decimos entonces que $X$ tiene una **distribución binomial** con parámetros $n$ y $p$, y escribimos:

$$
X\sim Bin(n,p)
$$

Esta expresión para la función de probabilidad puede obtenerse de la siguiente forma:

La probabilidad de obtener $x$ éxitos y $n-x$ fracasos en $n$ ensayos Bernoulli es, preliminarmente:

$$
\underbrace{p\cdot p\cdots p}_{x}
\underbrace{(1-p)(1-p)\cdots(1-p)}_{n-x}
$$

por lo tanto:

$$
=p^x(1-p)^{n-x}
$$

Sin embargo, en esta expresión se han colocado los $x$ éxitos en los primeros ensayos, cuando esto no necesariamente ocurrirá así.

Las diferentes formas en que los $x$ éxitos pueden distribuirse en los $n$ ensayos están dadas por el coeficiente binomial:

$$
\binom{n}{x}
$$

Por ejemplo:

$$
\binom{n}{n}=1
$$

representa la única manera de obtener $n$ éxitos en $n$ ensayos.

Además:

$$
\binom{n}{n-1}=n
$$

representa las diferentes formas de obtener $n-1$ éxitos en $n$ ensayos.

Al multiplicar este coeficiente binomial por el término:

$$
p^x(1-p)^{n-x}
$$

se obtiene la expresión final de la función de probabilidad para la distribución binomial:

$$
f(x)=\binom{n}{x}p^x(1-p)^{n-x}.
$$

Su esperanza y varianza estan dadas por:

$$E[X] = np \qquad\qquad \text{Var}(X) = np(1-p)$$

### Ejercicio 1

```python
n, p = 20, 0.05
dist_binomial = stats.binom(n, p)

x_vals = np.arange(0, n + 1)
print("Esperanza teorica E[X] =", dist_binomial.mean(), " (formula: n*p =", n * p, ")")
print("Varianza teorica Var(X) =", round(dist_binomial.var(), 4), " (formula: n*p*(1-p) =", round(n * p * (1 - p), 4), ")")
print("P(X = 0) (ningun defectuoso) =", round(dist_binomial.pmf(0), 4))
print("P(X <= 2) (a lo mas 2 defectuosos) =", round(dist_binomial.cdf(2), 4))
```

```python
plt.figure()
plt.bar(x_vals, dist_binomial.pmf(x_vals), color="tab:blue")
plt.xlabel("Numero de articulos defectuosos (x)")
plt.ylabel("P(X = x)")
plt.title(f"Binomial: numero de defectuosos en un lote de {n} (p = {p})")
plt.show()
```

### Ejercicio adicional

```python
p_mas_de_3 = 1 - dist_binomial.cdf(3)
print("P(X > 3) =", round(p_mas_de_3, 4))
```

## 3.7 Distribución Geométrica

Supongamos ahora que tenemos una sucesión infinita de ensayos independientes Bernoulli, en cada uno de los cuales la probabilidad de éxito es $p$.

Para cada una de estas sucesiones definimos la variable aleatoria $X$ como el número de fracasos antes de obtener el primer éxito.

Por ejemplo:

$$
X(EFFFEEE\cdots)=0
$$

$$
X(FEFEFF\cdots)=1
$$

$$
X(FFEEFF\cdots)=2
$$

$$
X(FFFEFE\cdots)=3
$$

$$
\vdots
$$

De esta manera, $X$ puede tomar los valores:

$$
0,1,2,\dots
$$

La probabilidad de que $X$ tome el valor entero $x\geq0$ es:

$$
(1-p)^xp
$$

ya que para obtener el primer éxito después de $x$ fracasos se requiere que ocurran:

- $x$ fracasos consecutivos con probabilidad $(1-p)^x$.
- Un éxito con probabilidad $p$.

Por lo tanto:

$$
P(X=x)=(1-p)^xp
$$

Decimos entonces que $X$ tiene una **distribución geométrica** con parámetro $p$ y escribimos:

$$
X\sim Geo(p)
$$

cuando su función de probabilidad está dada por:

$$
f(x)=
\begin{cases}
p(1-p)^x, & \text{si } x=0,1,\dots,\\
0, & \text{en otro caso}.
\end{cases}
$$

La esperanza de una variable aleatoria geométrica está dada por:

$$
E(X)=\frac{1-p}{p}
$$

Esta expresión representa el número promedio esperado de fracasos antes de obtener el primer éxito.


La varianza de una variable aleatoria geométrica es:

$$
Var(X)=\frac{1-p}{p^2}
$$

donde:

- $p$ es la probabilidad de éxito en cada ensayo.
- $X$ representa el número de fracasos antes del primer éxito.

### Ejercicio 1

```python
p = 0.2
dist_geometrica = stats.geom(p)

x_vals = np.arange(1, 21)
print("Esperanza teorica E[X] =", dist_geometrica.mean(), " (formula: 1/p =", 1 / p, ")")
print("Varianza teorica Var(X) =", round(dist_geometrica.var(), 4), " (formula: (1-p)/p^2 =", round((1 - p) / p**2, 4), ")")
print("P(X = 1) (venta en la primera llamada) =", round(dist_geometrica.pmf(1), 4))
```

```python
plt.figure()
plt.bar(x_vals, dist_geometrica.pmf(x_vals), color="tab:blue")
plt.xlabel("Numero de llamadas hasta la primera venta (x)")
plt.ylabel("P(X = x)")
plt.title("Geometrica: numero de llamadas hasta la primera venta (p = 0.2)")
plt.show()
```

```python
p_mas_de_5 = 1 - dist_geometrica.cdf(5)
print("P(X > 5) =", round(p_mas_de_5, 4))
```

## 3.8 Distribución Poisson

Supongamos que deseamos observar el número de ocurrencias de un cierto evento dentro de un intervalo de tiempo dado, por ejemplo, el número de clientes que llegan a un cajero automático durante la noche, el número de accidentes que ocurren en cierta avenida durante un día o el número de reclamaciones que llegan a una compañía aseguradora en una semana.

Para modelar este tipo de situaciones podemos definir la variable aleatoria $X$ como el número de ocurrencias de este evento en el intervalo de tiempo dado.

Es claro entonces que $X$ puede tomar los valores:

$$
0,1,2,\dots
$$

y, en principio, no existe una cota superior para el número de ocurrencias del evento.

Adicionalmente, supongamos que conocemos la **tasa media de ocurrencia** del evento de interés, la cual denotamos por la letra $\lambda$ (lambda).

El parámetro $\lambda$ es positivo y se interpreta como el número promedio de ocurrencias del evento por unidad de tiempo o espacio.

Decimos que $X$ tiene una **distribución Poisson** con parámetro $\lambda>0$, y escribimos:

$$
X\sim Poisson(\lambda)
$$

cuando su función de probabilidad está dada por:

$$
f(x)=
\begin{cases}
\dfrac{e^{-\lambda}\lambda^x}{x!}, & \text{si } x=0,1,2,\dots,\\
0, & \text{en otro caso}.
\end{cases}
$$

## Propiedades de la distribución Poisson

La distribución de Poisson se utiliza para modelar el número de veces que ocurre un evento en un intervalo fijo de tiempo o espacio, suponiendo que los eventos ocurren de manera independiente y con una tasa promedio constante.

La esperanza de una variable aleatoria con distribución Poisson es:

$$
E(X)=\lambda
$$

Es decir, el número promedio esperado de ocurrencias del evento es igual al parámetro $\lambda$.

La varianza de una variable aleatoria con distribución Poisson es:

$$
Var(X)=\lambda
$$

Por lo tanto, en una distribución Poisson la media y la varianza son iguales.

donde:

- $\lambda>0$ representa la tasa promedio de ocurrencia del evento.
- $X$ representa el número de ocurrencias observadas en un intervalo fijo de tiempo o espacio.

## 3.9 Distribución Binomial Negativa

Consideremos nuevamente la situación de observar los resultados de una sucesión infinita de ensayos independientes Bernoulli, en cada uno de los cuales la probabilidad de éxito es $p$.

Sea $r\geq1$ un número entero.

Definimos ahora a la variable aleatoria $X$ como el número de fracasos antes de obtener el $r$-ésimo éxito.

Es decir, la variable aleatoria cuenta cuántos fracasos ocurren antes de alcanzar un número fijo de éxitos.

Decimos entonces que $X$ tiene una **distribución binomial negativa** con parámetros $r$ y $p$, y escribimos:

$$
X\sim BinNeg(r,p)
$$

Es claro que la variable aleatoria $X$ puede tomar los valores:

$$
0,1,2,\dots
$$

con las probabilidades dadas por la función de probabilidad:

$$
f(x)=
\begin{cases}
\displaystyle\binom{r+x-1}{x}p^r(1-p)^x, & \text{si } x=0,1,2,\dots,\\
0, & \text{en otro caso}.
\end{cases}
$$

## Propiedades de la distribución Binomial Negativa

La distribución binomial negativa modela el número de fracasos que ocurren antes de obtener un número fijo $r$ de éxitos en una sucesión de ensayos Bernoulli independientes.

La esperanza de una variable aleatoria con distribución binomial negativa es:

$$
E(X)=\frac{r(1-p)}{p}
$$

La varianza está dada por:

$$
Var(X)=\frac{r(1-p)}{p^2}
$$

donde:

- $r$ representa el número de éxitos que se desea alcanzar.
- $p$ representa la probabilidad de éxito en cada ensayo.
- $X$ representa el número de fracasos observados antes de obtener el $r$-ésimo éxito.

## 3.10 Distribución Hipergeométrica

Supongamos que una población finita está formada por $N$ elementos, de los cuales $K$ poseen una determinada característica de interés y los $N-K$ restantes no la poseen.

Se selecciona una muestra de tamaño $n$ **sin reemplazo**, y se define la variable aleatoria $X$ como el número de elementos de la muestra que poseen dicha característica.

Decimos entonces que $X$ tiene una **distribución hipergeométrica** con parámetros $N$, $K$ y $n$, y escribimos:

$$
X\sim H(N,K,n).
$$

La variable aleatoria $X$ puede tomar los valores:

$$
x=\max(0,n+K-N),\ldots,\min(n,K).
$$

Su función de probabilidad está dada por:

$$
f(x)=
\begin{cases}
\displaystyle
\frac{\binom{K}{x}\binom{N-K}{\,n-x\,}}
{\binom{N}{n}},
& \text{si } x=\max(0,n+K-N),\ldots,\min(n,K),\\
0, & \text{en otro caso}.
\end{cases}
$$

La distribución hipergeométrica se utiliza cuando el muestreo se realiza **sin reemplazo**, por lo que las observaciones no son independientes.

## Propiedades de la distribución Hipergeométrica

La esperanza de una variable aleatoria con distribución hipergeométrica es:

$$
E(X)=n\frac{K}{N}.
$$

La varianza está dada por:

$$
Var(X)=
n\frac{K}{N}
\left(1-\frac{K}{N}\right)
\left(\frac{N-n}{N-1}\right).
$$

donde:

- $N$ es el tamaño de la población.
- $K$ es el número de elementos de la población que poseen la característica de interés.
- $n$ es el tamaño de la muestra seleccionada.
- $X$ representa el número de elementos con la característica de interés presentes en la muestra.

Una diferencia importante entre la distribución hipergeométrica y la distribución binomial es que la hipergeométrica considera muestreo **sin reemplazo**, mientras que la binomial supone ensayos independientes con reemplazo o una población suficientemente grande para considerar independencia entre observaciones.

### Ejercicio 1

```python
N, K, n = 50, 5, 10
dist_hipergeometrica = stats.hypergeom(N, K, n)

x_vals = np.arange(0, min(K, n) + 1)
esperanza_teorica = n * K / N
varianza_teorica = n * (K / N) * ((N - K) / N) * ((N - n) / (N - 1))

print("Esperanza teorica E[X] =", round(esperanza_teorica, 4))
print("Varianza teorica Var(X) =", round(varianza_teorica, 4))
print("P(X = 0) (ningun defectuoso en la muestra) =", round(dist_hipergeometrica.pmf(0), 4))
```

```python
plt.figure()
plt.bar(x_vals, dist_hipergeometrica.pmf(x_vals), color="tab:blue")
plt.xlabel("Numero de defectuosos en la muestra (x)")
plt.ylabel("P(X = x)")
plt.title("Hipergeometrica: defectuosos en una muestra sin reemplazo (N=50, K=5, n=10)")
plt.show()
```

# 4. Continuas

Una variable aleatoria continua es aquella que puede tomar cualquier valor dentro de uno o varios intervalos de los números reales.

A diferencia de las variables aleatorias discretas, una variable continua no asigna probabilidades a valores individuales, sino a intervalos de valores.

La probabilidad de que una variable aleatoria continua tome exactamente un valor es igual a cero, es decir,

$$
P(X=x)=0.
$$

Las probabilidades se obtienen mediante una **función de densidad**, la cual cumple que:

$$
P(a\leq X\leq b)
=
\int_a^b f(x)\,dx.
$$

Además, toda función de densidad debe satisfacer:

$$
f(x)\geq0
$$

y

$$
\int_{-\infty}^{\infty}f(x)\,dx=1.
$$


## 4.1 Función de densidad

En una variable aleatoria continua, las probabilidades no se asignan a valores individuales, sino a intervalos de valores. Para ello se utiliza la **función de densidad de probabilidad**, denotada por $f(x)$.

Una función $f(x)$ es una función de densidad si cumple las siguientes propiedades:

1. Es no negativa para todo valor de $x$:

$$
f(x)\geq0.
$$

2. El área total bajo la curva es igual a uno:

$$
\int_{-\infty}^{\infty}f(x)\,dx=1.
$$

La probabilidad de que la variable aleatoria tome un valor dentro del intervalo $[a,b]$ se obtiene mediante la integral:

$$
P(a\leq X\leq b)=\int_a^b f(x)\,dx.
$$

Además, para cualquier variable aleatoria continua se cumple que:

$$
P(X=x)=0.
$$


## 4.2 Función de distribución

La **función de distribución acumulada**, denotada por $F(x)$, representa la probabilidad de que la variable aleatoria tome un valor menor o igual que $x$.

Se define como:

$$
F(x)=P(X\leq x).
$$

Para una variable aleatoria continua, esta función puede obtenerse integrando la función de densidad:

$$
F(x)=\int_{-\infty}^{x}f(t)\,dt.
$$

La función de distribución cumple las siguientes propiedades:

- $0\leq F(x)\leq1$.
- Es una función creciente.
- $\displaystyle\lim_{x\to-\infty}F(x)=0$.
- $\displaystyle\lim_{x\to\infty}F(x)=1$.



## 4.3 Valor esperado de una variable aleatoria continua (Esperanza)

El **valor esperado** o **esperanza matemática** de una variable aleatoria continua representa el valor promedio que se espera obtener después de repetir el experimento un gran número de veces.

Se define como:

$$
E(X)=\int_{-\infty}^{\infty}x\,f(x)\,dx.
$$


## 4.4 Varianza

La **varianza** mide la dispersión de los valores de la variable aleatoria respecto a su media.

Se define como:

$$
Var(X)=E[(X-E(X))^2].
$$

De forma equivalente, puede calcularse mediante:

$$
Var(X)=E(X^2)-[E(X)]^2,
$$

donde

$$
E(X^2)=\int_{-\infty}^{\infty}x^2f(x)\,dx.
$$

La desviación estándar es la raíz cuadrada positiva de la varianza:

$$
\sigma=\sqrt{Var(X)}.
$$

## 4.5 Distribución Uniforme Continua

Una variable aleatoria tiene una distribución uniforme continua cuando todos los valores dentro del intervalo $[a,b]$ tienen la misma densidad.

Se escribe:

$$
X\sim U(a,b).
$$

## Función de densidad

$$
f(x)=
\begin{cases}
\dfrac{1}{b-a},
&
a\leq x\leq b,\\
0,
&
\text{en otro caso.}
\end{cases}
$$

## Función de distribución

$$
F(x)=
\begin{cases}
0,
&
x<a,\\
\dfrac{x-a}{b-a},
&
a\leq x\leq b,\\
1,
&
x>b.
\end{cases}
$$

## Esperanza

$$
E(X)=\frac{a+b}{2}.
$$

## Varianza

$$
Var(X)=\frac{(b-a)^2}{12}.
$$

### Ejercicio 1

```python
a, b = 0, 10
dist_unif_cont = stats.uniform(loc=a, scale=b - a)

print("Esperanza teorica E[X] =", dist_unif_cont.mean(), " (formula: (a+b)/2 =", (a + b) / 2, ")")
print("Varianza teorica Var(X) =", round(dist_unif_cont.var(), 4), " (formula: (b-a)^2/12 =", round((b - a) ** 2 / 12, 4), ")")
print("P(X <= 3) =", round(dist_unif_cont.cdf(3), 4))
```

## 4.6 Distribución Normal

La distribución normal es una de las distribuciones continuas más importantes en probabilidad y estadística.

Muchas variables naturales presentan un comportamiento aproximadamente normal.

Se escribe:

$$
X\sim N(\mu,\sigma^2).
$$

donde

- $\mu$ representa la media.
- $\sigma^2$ representa la varianza.

## Función de densidad

$$
f(x)=
\frac{1}{\sigma\sqrt{2\pi}}
e^{-\frac{(x-\mu)^2}{2\sigma^2}}.
$$

## Función de distribución

La función de distribución se define como:

$$
F(x)=
P(X\leq x)
=
\int_{-\infty}^{x}
f(t)\,dt.
$$

Esta integral no posee una solución analítica sencilla y normalmente se calcula mediante tablas o software estadístico.

## Esperanza

$$
E(X)=\mu.
$$

## Varianza

$$
Var(X)=\sigma^2.
$$

### Ejercicio 1

```python
mu, sigma = 170, 8
dist_normal = stats.norm(loc=mu, scale=sigma)

print("Esperanza teorica E[X] =", dist_normal.mean(), " (formula: mu =", mu, ")")
print("Varianza teorica Var(X) =", dist_normal.var(), " (formula: sigma^2 =", sigma**2, ")")
print("P(X <= 180) =", round(dist_normal.cdf(180), 4))
print("P(160 <= X <= 180) =", round(dist_normal.cdf(180) - dist_normal.cdf(160), 4))
```

```python
x_vals = np.linspace(mu - 4 * sigma, mu + 4 * sigma, 400)
fig, axes = plt.subplots(1, 2, figsize=(11, 4.5))
axes[0].plot(x_vals, dist_normal.pdf(x_vals), label="f(x)")
axes[0].set_title("Normal: densidad")
axes[0].set_xlabel("Estatura (cm)")
axes[0].set_ylabel("f(x)")
axes[0].legend()

axes[1].plot(x_vals, dist_normal.cdf(x_vals), color="tab:orange", label="F(x)")
axes[1].set_title("Normal: distribucion acumulada")
axes[1].set_xlabel("Estatura (cm)")
axes[1].set_ylabel("F(x)")
axes[1].legend()
plt.tight_layout()
plt.show()
```

## 4.8 Distribución Gamma

La variable aleatoria continua $X$ tiene una **distribución Gamma** con parámetros $\alpha>0$ y $\lambda>0$, y se escribe:

$$
X\sim Gamma(\alpha,\lambda).
$$

La distribución Gamma se utiliza para modelar tiempos de espera hasta que ocurren varios eventos independientes. Es una generalización de la distribución exponencial, la cual se obtiene cuando $\alpha=1$.

## Función de densidad

La función de densidad está dada por:

$$
f(x)=
\begin{cases}
\displaystyle
\frac{\lambda(\lambda x)^{\alpha-1}}{\Gamma(\alpha)}
e^{-\lambda x},
&
x>0,\\
0,
&
\text{en otro caso.}
\end{cases}
$$

donde $\Gamma(\alpha)$ es la **función Gamma**, definida por:

$$
\Gamma(\alpha)=
\int_{0}^{\infty}
x^{\alpha-1}e^{-x}\,dx.
$$

Cuando $\alpha$ es un número entero positivo, se cumple que:

$$
\Gamma(\alpha)=(\alpha-1)!.
$$

## Función de distribución

La función de distribución se define como:

$$
F(x)=
\int_0^x
f(t)\,dt.
$$

En general, esta integral no posee una expresión cerrada sencilla y se calcula mediante la **función Gamma incompleta**, por lo que normalmente se evalúa utilizando software estadístico.

## Esperanza

La esperanza de una variable aleatoria con distribución Gamma es:

$$
E(X)=\frac{\alpha}{\lambda}.
$$

## Varianza

La varianza está dada por:

$$
Var(X)=\frac{\alpha}{\lambda^2}.
$$

donde:

- $\alpha$ es el parámetro de forma.
- $\lambda$ es el parámetro de tasa.
- $X$ representa una variable aleatoria con distribución Gamma.

### Ejercicio 1

```python
k, lam = 4, 3
dist_gamma = stats.gamma(a=k, scale=1 / lam)  # scipy usa 'a' como el parametro de forma k

print("Esperanza teorica E[X] =", round(dist_gamma.mean(), 4), " (formula: k/lambda =", round(k / lam, 4), ")")
print("Varianza teorica Var(X) =", round(dist_gamma.var(), 4), " (formula: k/lambda^2 =", round(k / lam**2, 4), ")")
print("P(X <= 1) (menos de 1 hora para 4 clientes) =", round(dist_gamma.cdf(1), 4))
```

```python
x_vals = np.linspace(0, 4, 400)
fig, axes = plt.subplots(1, 2, figsize=(11, 4.5))
axes[0].plot(x_vals, dist_gamma.pdf(x_vals), label="f(x)")
axes[0].set_title("Gamma: densidad")
axes[0].set_xlabel("Tiempo hasta el 4o cliente (horas)")
axes[0].set_ylabel("f(x)")
axes[0].legend()

axes[1].plot(x_vals, dist_gamma.cdf(x_vals), color="tab:orange", label="F(x)")
axes[1].set_title("Gamma: distribucion acumulada")
axes[1].set_xlabel("Tiempo hasta el 4o cliente (horas)")
axes[1].set_ylabel("F(x)")
axes[1].legend()
plt.tight_layout()
plt.show()
```

# 5. Simulación de variables aleatorias: método de la transformada inversa

Si $U \sim \text{Uniforme}(0,1)$ y $F$ es la función de distribución (continua y estrictamente creciente) de una variable $X$, entonces

$$X = F^{-1}(U)$$

tiene exactamente la distribución $F$. Este resultado, conocido como **método de la transformada inversa**, permite simular cualquier variable aleatoria de la que se conozca $F^{-1}$ en forma cerrada a partir de números uniformes.

Para la exponencial de tasa $\lambda$, $F(x) = 1 - e^{-\lambda x}$, por lo que

$$F^{-1}(u) = -\frac{\ln(1-u)}{\lambda}.$$

Se implementa manualmente esta fórmula (sin usar `dist.rvs`) y se compara contra la teoría.

```python
lam = 3
u = rng.uniform(0, 1, size=10_000)
muestra_transformada_inversa = -np.log(1 - u) / lam  # F^-1(u) para la exponencial

media_teorica = 1 / lam
var_teorica = 1 / lam**2
media_muestral = muestra_transformada_inversa.mean()
var_muestral = muestra_transformada_inversa.var()

print("Media teorica:", round(media_teorica, 4), " | Media muestral (transformada inversa):", round(media_muestral, 4))
print("Varianza teorica:", round(var_teorica, 4), " | Varianza muestral (transformada inversa):", round(var_muestral, 4))
```

```python
x_vals = np.linspace(0, 2, 400)
plt.figure()
plt.hist(muestra_transformada_inversa, bins=40, density=True, alpha=0.6, color="tab:blue",
         label="Simulada (transformada inversa)")
plt.plot(x_vals, stats.expon(scale=1 / lam).pdf(x_vals), color="tab:red", label="Densidad teorica")
plt.xlabel("x")
plt.ylabel("Densidad")
plt.title("Transformada inversa: exponencial simulada vs. teoria")
plt.legend()
plt.show()
```

# 6. Función generadora de momentos (FGM)

Otra función de gran utilidad en probabilidad y estadística, aplicable tanto a variables aleatorias discretas como continuas, es la **función generadora de momentos (FGM)**. Esta función está relacionada con los momentos de una variable aleatoria, como la esperanza, la varianza y otros momentos de orden superior.

### Definición 2.14

La **función generadora de momentos** de una variable aleatoria discreta o continua $X$ es la función $M(t)$ definida como:

$$
M(t)=E\left(e^{tX}\right),
$$

para aquellos valores reales de $t$ en los que dicha esperanza exista.

En el caso de una variable aleatoria discreta, la función generadora de momentos puede calcularse mediante:

$$
M(t)=\sum_x e^{tx}P(X=x).
$$

Para una variable aleatoria continua, la definición equivalente es:

$$
M(t)=\int_{-\infty}^{\infty}e^{tx}f(x)\,dx,
$$

donde $f(x)$ es la función de densidad de la variable aleatoria.

La función generadora de momentos recibe este nombre porque, cuando existe, permite obtener los momentos de la variable aleatoria derivando la función respecto de $t$ y evaluando en $t=0$. En particular:

$$
E(X)=M'(0),
$$

y

$$
Var(X)=M''(0)-[M'(0)]^2.
$$

La FGM es una herramienta muy útil para identificar distribuciones de probabilidad, demostrar resultados teóricos y calcular momentos de manera más sencilla.

```python
lam = 3

def fgm_exponencial(t, lam=lam):
    return lam / (lam - t)

def derivada_numerica(func, t0, h=1e-5, orden=1):
    """Aproxima la derivada de 'func' en t0 usando diferencias finitas centradas."""
    if orden == 1:
        return (func(t0 + h) - func(t0 - h)) / (2 * h)
    if orden == 2:
        return (func(t0 + h) - 2 * func(t0) + func(t0 - h)) / h**2
    raise ValueError("Solo se implementan las derivadas de orden 1 y 2")

E_X_fgm = derivada_numerica(fgm_exponencial, 0, orden=1)
E_X2_fgm = derivada_numerica(fgm_exponencial, 0, orden=2)

dist_exponencial = stats.expon(scale=1 / lam)
print("M_X'(0)  (derivada numerica) =", round(E_X_fgm, 4), " | E[X] teorica (scipy) =", round(dist_exponencial.mean(), 4))
print("M_X''(0) (derivada numerica) =", round(E_X2_fgm, 4), " | E[X^2] teorica =",
      round(dist_exponencial.var() + dist_exponencial.mean()**2, 4))
```

# 7. Función característica (FC)

La función característica de $X$ se define como

$$\varphi_X(t) = E\left[e^{itX}\right] = E[\cos(tX)] + i\,E[\sin(tX)],$$

donde $i = \sqrt{-1}$. A diferencia de la FGM, **siempre existe** (porque $|e^{itX}| = 1$), y también caracteriza de forma única a la distribución.

**Ejemplo (Normal):** para $X \sim N(\mu, \sigma^2)$, la forma cerrada es

$$\varphi_X(t) = e^{i\mu t - \tfrac{1}{2}\sigma^2 t^2}.$$

Se verifica numéricamente promediando $e^{itX}$ sobre una muestra simulada y comparando con la fórmula cerrada.

```python
mu, sigma = 170, 8
dist_normal = stats.norm(loc=mu, scale=sigma)
muestra = dist_normal.rvs(size=200_000, random_state=rng)

t_vals = np.linspace(-0.5, 0.5, 100)
fc_teorica = np.exp(1j * mu * t_vals - 0.5 * sigma**2 * t_vals**2)
fc_empirica = np.array([np.mean(np.exp(1j * t * muestra)) for t in t_vals])

print("phi_X(0.1) teorica :", np.round(fc_teorica[np.argmin(np.abs(t_vals - 0.1))], 4))
print("phi_X(0.1) empirica:", np.round(fc_empirica[np.argmin(np.abs(t_vals - 0.1))], 4))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4.5))
axes[0].plot(t_vals, fc_teorica.real, label="Teorica", linewidth=2)
axes[0].plot(t_vals, fc_empirica.real, "--", label="Empirica")
axes[0].set_title("Funcion caracteristica: parte real")
axes[0].set_xlabel("t")
axes[0].set_ylabel("Re(phi_X(t))")
axes[0].legend()

axes[1].plot(t_vals, fc_teorica.imag, label="Teorica", linewidth=2)
axes[1].plot(t_vals, fc_empirica.imag, "--", label="Empirica")
axes[1].set_title("Funcion caracteristica: parte imaginaria")
axes[1].set_xlabel("t")
axes[1].set_ylabel("Im(phi_X(t))")
axes[1].legend()
plt.tight_layout()
plt.show()
```

# 8. Ley de los grandes números

El teorema conocido como la ley de los grandes números es un resultado muy interesante que puede observarse en la naturaleza. Constituye uno de los resultados más importantes de la teoría de la probabilidad y tiene mucha relevancia en las aplicaciones tanto téoricas como prácticas. Este teorema establece que, bajo ciertas condiciones, el promedio aritmético de variables aleatorias converge a una constante cuando el número de sumandos crece a infinito.

**Definición:**

Sea $X_1,X_2,\ldots$ una sucesión infinita de variables aleatorias independientes e idénticamente distribuidas con media finita $\mu$. Entonces, cuando $n\rightarrow\infty$,

$$
\frac{1}{n}\sum_{i=1}^{n}X_i
\longrightarrow
\mu.
$$

donde la convergencia se verifica en el sentido **casi seguro** (Ley Fuerte de los Grandes Números) y también **en probabilidad** (Ley Débil de los Grandes Números).

```python
lam = 3
n_max = 20_000
muestra = stats.expon(scale=1 / lam).rvs(size=n_max, random_state=rng)

tamanos = np.arange(1, n_max + 1)
promedio_acumulado = np.cumsum(muestra) / tamanos

print("Esperanza teorica E[X] =", round(1 / lam, 4))
print("Promedio muestral con n=10:   ", round(promedio_acumulado[9], 4))
print("Promedio muestral con n=1000: ", round(promedio_acumulado[999], 4))
print(f"Promedio muestral con n={n_max}:", round(promedio_acumulado[-1], 4))
```

```python
plt.figure()
plt.plot(tamanos, promedio_acumulado, label="Promedio muestral acumulado")
plt.axhline(1 / lam, color="red", linestyle="--", label="Esperanza teorica E[X] = 1/lambda")
plt.xscale("log")
plt.xlabel("Tamano de muestra n (escala logaritmica)")
plt.ylabel("Promedio muestral")
plt.title("Ley de los Grandes Numeros: convergencia del promedio muestral")
plt.legend()
plt.show()
```

# 9. Teorema Central del Límite

El **Teorema Central del Límite (TCL)** es uno de los resultados más importantes de la teoría de la probabilidad y la estadística. Tiene una amplia gama de aplicaciones y permite aproximar diversas distribuciones mediante la distribución normal.

En particular, este teorema facilita el cálculo de probabilidades y constituye la base de muchos métodos de inferencia estadística.


Sea $X_1,X_2,\ldots$ una sucesión infinita de variables aleatorias independientes e idénticamente distribuidas, con media $\mu$ y varianza finita $\sigma^2$.

Entonces, la función de distribución de la variable aleatoria

$$
Z_n=
\frac{(X_1+X_2+\cdots+X_n)-n\mu}
{\sqrt{n\sigma^2}}
$$

tiende a la función de distribución de una **distribución normal estándar** cuando $n$ tiende a infinito.

Es decir,

$$
Z_n
\overset{d}{\longrightarrow}
N(0,1),
\qquad
\text{cuando }
n\rightarrow\infty.
$$

```python
lam = 3
mu, sigma = 1 / lam, 1 / lam  # media y desviacion estandar de la exponencial(lambda=3)
tamanos_muestra = [1, 5, 30, 100]
n_repeticiones = 10_000

medias_por_n = {}
for n in tamanos_muestra:
    datos = stats.expon(scale=1 / lam).rvs(size=(n_repeticiones, n), random_state=rng)
    medias_por_n[n] = datos.mean(axis=1)
    print(f"n={n:>3}: media de las medias muestrales = {medias_por_n[n].mean():.4f}, "
          f"desviacion estandar = {medias_por_n[n].std():.4f} "
          f"(teorica: {sigma / np.sqrt(n):.4f})")
```

```python
fig, axes = plt.subplots(2, 2, figsize=(11, 8))
for ax, n in zip(axes.ravel(), tamanos_muestra):
    ax.hist(medias_por_n[n], bins=40, density=True, alpha=0.6, color="tab:blue", label="Medias muestrales")
    x_vals = np.linspace(medias_por_n[n].min(), medias_por_n[n].max(), 200)
    ax.plot(x_vals, stats.norm(loc=mu, scale=sigma / np.sqrt(n)).pdf(x_vals), color="tab:red",
            label="Normal teorica (TCL)")
    ax.set_title(f"n = {n}")
    ax.set_xlabel("Media muestral")
    ax.set_ylabel("Densidad")
    ax.legend()
plt.suptitle("Teorema Central del Limite: distribucion de la media muestral (poblacion exponencial)")
plt.tight_layout()
plt.show()
```

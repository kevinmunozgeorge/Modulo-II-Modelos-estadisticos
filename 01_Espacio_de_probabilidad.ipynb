# 1. Espacio de Probabilidad

Un espacio de probabilidad es una terna $\{\Omega, \mathscr{F}, \mathbb{P}\}$ en
donde $\Omega$ es un conjunto arbitrario, $\mathscr{F}$ es una $σ$-álgebra de subconjuntos de $\Omega$ y $\mathbb{P}$ es una medida de probabilidad definida sobre $\mathscr{F}$.

En los siguientes puntos de este notebook vamos a definir con mayor detalle cada uno de los elementos que conforman nuestro espacio de probabilidad, brindando además ejemplos.


# 2. Espacio Muestral

Comencemos explorando el **espacio muestral**, también llamado espacio muestra, el espacio muestral de un experimento aleatorio es el conjunto de todos los posibles resultados del experimento y se le denota, generalmente, por la letra griega
$\Omega$ (omega mayúscula). A un resultado particular del experimento se le denota por la letra $\omega$ (omega minúscula).

### Ejemplo 1.

Si un experimento aleatorio consiste en lanzar un dado y
observar el número que aparece en la cara superior, entonces claramente
el espacio muestral es el conjunto $\Omega=\{1,2,3,4,5,6\}$. Como ejemplo de
un evento para este experimento podemos definir el conjunto $A=\{2,4,6\}$, que corresponde al suceso de obtener como resultado un número par. Si al
lanzar el dado una vez se obtiene el número “4”, decimos entonces que se
observó la ocurrencia del evento A, y si se obtiene, por ejemplo, el resultado
“1”, decimos que no se observó la ocurrencia del evento A.

Veamos dicho ejemplo elaborado con el siguiente código, el cual podemos ejecutar tantas veces queramos con el fin de ver cuantas veces obtenemos el evento $A$

```python
# ============================================
# Ejemplo 1.Lanzamiento de un dado
# ============================================

import random

# Espacio muestral
omega = {1, 2, 3, 4, 5, 6}

# Evento A: obtener un número par
A = {2, 4, 6}

# Simulación de un lanzamiento de dado
resultado = random.choice(list(omega))

print("Espacio muestral (Ω):", omega)
print("Evento A (números pares):", A)
print("Resultado obtenido:", resultado)

# Verificar si ocurrió el evento A
if resultado in A:
    print("✅ Se observó la ocurrencia del evento A.")
else:
    print("❌ No se observó la ocurrencia del evento A.")
```

# 3. Eventos

En teoría de la probabilidad, un **evento** es cualquier subconjunto del espacio muestral $\Omega$. Es decir, un evento está formado por uno o más resultados posibles de un experimento aleatorio.

Los eventos suelen representarse con letras mayúsculas, como $A, B$ o $C$. Un evento ocurre cuando el resultado obtenido en el experimento pertenece al conjunto que lo define.

Retomando nuestro ejemplo 1 del apartado anterior; al lanzar un dado equilibrado, ya sabemos que el espacio muestral es

$\Omega=\{1,2,3,4,5,6\}.$

Se pueden definir los siguientes eventos:

$A$: obtener un número par, es decir;

$A=\{2,4,6\}$

$B$: obtener un número mayor que 3;

$B=\{4,5,6\}$

Después de realizar el experimento, se verifica si el resultado obtenido pertenece o no al evento considerado.

# 4. Álgrebras y sigma-álgebras

## álgebra de eventos

Una colección $\mathscr{F}$ de subconjuntos de un espacio muestral $\Omega$ es una **álgebra** si cumple las tres condiciones siguientes:

1. $\Omega \in \mathscr{F}$

2. Si $A \in \mathscr{F}$, entonces $A^{c} \in \mathscr{F}$

3. Si $A_{1}, A_{2}, \ldots, A_{n} \in \mathscr{F}$, entonces

$$
\bigcup_{k=1}^{n} A_{k} \in \mathscr{F}.
$$

Cuando esta tercera condición es también válida para sucesiones infinitas de eventos, a la correspondiente colección de subconjuntos de $\Omega$ se le llama $\sigma$-álgebra (se lee sigma álgebra), en donde el prefijo $\sigma$ se refiere a la operación infinita involucrada.
La definición es entonces muy parecida a la anterior

## σ-álgebra de eventos

Una colección $\mathscr{F}$ de subconjuntos de un espacio muestral $\Omega$ es una **σ-álgebra** si cumple las tres condiciones siguientes:

1. $\Omega \in \mathscr{F}$

2. Si $A \in \mathscr{F}$, entonces $A^{c} \in \mathscr{F}$

3. Si $A_{1}, A_{2}, \ldots, A_{n} \in \mathscr{F}$, entonces

$$
\bigcup_{k=1}^{∞} A_{k} \in \mathscr{F}.
$$

A los elementos de $\mathscr{F}$ se les llama **eventos**.

# 5. Medida de probabilidad

Una **medida de probabilidad** es una función $\mathbb{P}$ que asigna a cada evento de una σ-álgebra un número real comprendido entre 0 y 1, el cual representa la posibilidad de que dicho evento ocurra.

Formalmente, sea $(\Omega,\mathscr{F})$ un espacio medible. Una medida de probabilidad es una función

$P:\mathscr{F}\rightarrow[0,1]$

que satisface las siguientes propiedades:

1. **No negatividad**

$ P(A)\geq 0,\qquad \forall A\in\mathscr{F}.$

2. **Normalización**

$P(\Omega)=1$

3. **σ-aditividad (aditividad numerable)**

Si $A_1,A_2,\ldots$ son eventos mutuamente excluyentes, es decir,

$A_i\cap A_j=\varnothing,\qquad i\neq j,$

entonces

$$P\left(\bigcup_{k=1}^{\infty}A_k\right)=\sum_{k=1}^{\infty}P(A_k)$$



Estas tres propiedades garantizan que las probabilidades asignadas a los eventos sean consistentes y constituyen los **axiomas de Kolmogorov**, sobre los cuales se fundamenta la teoría moderna de la probabilidad.

En resumen, a cualquier función $\mathbb{P}$ definida sobre una colección de
eventos que satisfaga los tres axiomas de Kolmogorov, se le llama medida
de probabilidad o, simplemente, probabilidad.

# 6. Propiedades de la probabilidad.

1. $\mathbb{P}(∅)=0$

2. Sea $A_{1}, A_{2}, \ldots, A_{n}$ una colección finita de eventos **ajenos dos a dos** (es decir, mutuamente excluyentes). Entonces,

$$
P\left(\bigcup_{k=1}^{n} A_{k}\right)
=
\sum_{k=1}^{n} P(A_{k}).
$$

3. Para cualquier evento $A$, $\mathbb{P}(A^c)=1-\mathbb{P}$

4. Si $A \subseteq B$ entonces $\mathbb{P}(A)\leq \mathbb{P}(B)$

5. Si $A \subseteq B$ entonces $P(B \setminus A) = P(B) - P(A)$

6. Para cualquier evento $A$, $0 \leq P(A) \leq 1$

7. Para cualesquiera eventos $A$ y $B$;
 $$P(A \cup B) = P(A) + P(B) - P(A \cap B)$$

8. Para cualesquiera eventos $A$, $B$ y $C$;
$$P(A \cup B \cup C) = P(A) + P(B) + P(C) - P(A \cap B) - P(A \cap C) - P(B \cap C) + P(A \cap B \cap C)$$

# 7. Probabilidad clásica

Con lo visto anteriormente, ya sabemos que la probabilidad de un evento A es un número real en el intervalo $\[0,1\]$ que
se denota por $\mathbb{P}$ y representa una medida de la frecuencia con la que se observa la ocurrencia de este evento cuando se efectúa el experimento aleatorio en cuestión.
Existen definiciones específicas de la probabilidad,
algunas de las cuales estudiaremos en las siguientes secciones.

Comencemos con la **definición clásica de probabilidad**:

Sea $A$ un subconjunto de un espacio muestral $\Omega$ de cardinalidad finita. La **probabilidad clásica** de un evento $A$ se define como el cociente entre el número de resultados favorables y el número total de resultados posibles, es decir,

$$
P(A)=\frac{\#A}{\#\Omega},
$$

donde:

- $\#A$ representa la **cardinalidad** del conjunto $A$, es decir, el número de elementos que contiene.
- $\#\Omega$ representa la **cardinalidad** del espacio muestral $\Omega$

La definición de probabilidad clásica es válida siempre que **todos los resultados del espacio muestral sean igualmente probables**.

### Ejemplo 1

Considérese nuevamente el experimento de lanzar un dado equilibrado de seis caras.

Ya sabemos que el espacio muestral es

$$
\Omega=\{1,2,3,4,5,6\}.
$$

Sea el evento

$$
A=\{2,3,5\},
$$

que representa el suceso **"obtener un número primo"**.

Como todos los resultados son igualmente probables, la probabilidad clásica del evento es

$$
P(A)=\frac{\#A}{\#\Omega}
=\frac{3}{6}
=\frac{1}{2}=0.5.
$$

Es decir, existe una probabilidad del **50 %** de obtener un número primo al lanzar un dado.

```python
# ============================================
# Ejemplo de probabilidad clásica
# ============================================

# Espacio muestral
omega = {1, 2, 3, 4, 5, 6}

# Evento: obtener un número primo
A = {2, 3, 5}

# Cálculo de la probabilidad clásica
P_A = len(A) / len(omega)

print("Espacio muestral (Ω):", omega)
print("Evento A (números primos):", A)
print("Número de casos favorables:", len(A))
print("Número total de casos posibles:", len(omega))
print(f"P(A) = {len(A)}/{len(omega)} = {P_A:.2f}")
```

### Ejemplo 2

Una urna contiene **20 pelotas**, de las cuales **5 son rojas**, **8 son azules** y **7 son verdes**. Se extrae una pelota al azar.

Sea el evento

$$
A=\text{"extraer una pelota roja"}.
$$

Como todas las pelotas tienen la misma probabilidad de ser seleccionadas, la probabilidad clásica del evento es

$$
P(A)=\frac{5}{20}=\frac{1}{4}=0.25.
$$

Esto significa que existe un **25 %** de probabilidad de extraer una pelota roja.

```python
# ============================================
# Ejemplo: extracción de una pelota de una urna
# ============================================

import random

# Construcción de la urna
urna = (
    ["Roja"] * 5 +
    ["Azul"] * 8 +
    ["Verde"] * 7
)

# Extraer una pelota al azar
pelota = random.choice(urna)

# Probabilidad clásica
favorables = urna.count("Roja")
total = len(urna)

probabilidad = favorables / total

print("Total de pelotas:", total)
print("Pelotas rojas:", favorables)
print(f"P(A) = {favorables}/{total} = {probabilidad:.2f}")

print("\nPelota extraída:", pelota)

if pelota == "Roja":
    print("✅ Ocurrió el evento A (extraer una pelota roja).")
else:
    print("❌ No ocurrió el evento A.")
```

# 8. Probabilidad Condicional

La **probabilidad condicional** es la probabilidad de que ocurra un evento $A$ sabiendo que previamente ha ocurrido otro evento $B$.

En otras palabras, cuando se conoce que el evento $B$ ha ocurrido, el espacio muestral se reduce únicamente a los resultados contenidos en dicho evento, y la probabilidad del evento $A$ debe calcularse considerando únicamente ese nuevo espacio.

La probabilidad condicional de $A$ dado $B$ se define como

$$
P(A\mid B)=\frac{P(A\cap B)}{P(B)},
$$

siempre que

$$
P(B)>0.
$$

donde:

- $P(A\mid B)$ es la probabilidad de que ocurra $A$ dado que ocurrió $B$.
- $P(A\cap B)$ es la probabilidad de que ocurran simultáneamente los eventos $A$ y $B$.
- $P(B)$ es la probabilidad del evento condicionante.

### Ejemplo 1

Considérese nuevamente nuestro experimento de lanzar un dado equilibrado.

Sea

$$
A=\{2,4,6\},
$$

el evento **"obtener un número par"**, y

$$
B=\{4,5,6\},
$$

el evento **"obtener un número mayor que 3"**.

La intersección de ambos eventos claramente es

$$
A\cap B=\{4,6\}.
$$

Por lo tanto,

$$
P(A\mid B)
=
\frac{P(A\cap B)}{P(B)}
=
\frac{2/6}{3/6}
=
\frac23
\approx0.667.
$$

Esto significa que, sabiendo que el resultado fue mayor que 3, la probabilidad de que además sea un número par es de aproximadamente **66.7 %**.

```python
# ============================================
# Ejemplo de probabilidad condicional
# ============================================

# Espacio muestral
omega = {1, 2, 3, 4, 5, 6}

# Definición de eventos
A = {2, 4, 6}      # Números pares
B = {4, 5, 6}      # Números mayores que 3

# Intersección
interseccion = A & B

# Probabilidades
P_B = len(B) / len(omega)
P_inter = len(interseccion) / len(omega)
P_condicional = P_inter / P_B

print("Espacio muestral (Ω):", omega)
print("Evento A:", A)
print("Evento B:", B)
print("A ∩ B:", interseccion)

print(f"\nP(B) = {P_B:.3f}")
print(f"P(A ∩ B) = {P_inter:.3f}")
print(f"P(A | B) = {P_condicional:.3f}")
```

# 9. Independencia

Sean dos eventos $A$ y $B$, decimos que estos son **independientes** si

$$P(A \cap B) = P(A) \cdot P(B).$$

En otras palabras, podemos decir que la independencia de eventos en probabilidad ocurre cuando el resultado de un primer suceso no cambia ni afecta la probabilidad de que ocurra un segundo suceso.

### Ejemplo 1

Considérese el experimento de lanzar **dos monedas equilibradas**.

Podemos construir nuestro espacio muestral como todos los posibles eventos a ocurrir, esto es:

$$
\Omega=
\{
(\text{Cara},\text{Cara}),
(\text{Cara},\text{Cruz}),
(\text{Cruz},\text{Cara}),
(\text{Cruz},\text{Cruz})
\}.
$$

Definamos los eventos:

- $A$: "La primera moneda cae en Cara."
- $B$: "La segunda moneda cae en Cara."

Las probabilidades son

$$
P(A)=\frac{2}{4}=\frac12,
$$

$$
P(B)=\frac{2}{4}=\frac12,
$$

y

$$
P(A\cap B)=\frac14.
$$

Como

$$
P(A)P(B)=\frac12\cdot\frac12=\frac14=P(A\cap B),
$$

se concluye que los eventos \(A\) y \(B\) son independientes.

### Ejemplo 2

Considérese ahora el experimento de lanzar **un dado equilibrado** y **una moneda equilibrada** al mismo tiempo.

El espacio muestral está formado por las 12 combinaciones posibles entre los resultados del dado y la moneda.

Definamos los siguientes eventos:

- $A$: "Obtener un número par en el dado."
- $B$: "Obtener cara en la moneda."

Las probabilidades son

$$
P(A)=\frac{3}{6}=\frac12
$$

$$
P(B)=\frac12
$$

y

$$
P(A\cap B)=\frac{3}{12}=\frac14
$$

Como

$$
P(A)\cdot P(B)=\frac12\cdot\frac12=\frac14=P(A\cap B),
$$

se concluye que los eventos \(A\) y \(B\) son independientes.

```python
# ============================================
# Independencia: lanzamiento de un dado y una moneda
# ============================================

import random

N = 1000000

evento_A = 0      # Número par
evento_B = 0      # Cara
interseccion = 0

for _ in range(N):

    dado = random.randint(1, 6)
    moneda = random.choice(["Cara", "Cruz"])

    if dado % 2 == 0:
        evento_A += 1

    if moneda == "Cara":
        evento_B += 1

    if dado % 2 == 0 and moneda == "Cara":
        interseccion += 1

P_A = evento_A / N
P_B = evento_B / N
P_AB = interseccion / N

print(f"P(A) = {P_A:.4f}")
print(f"P(B) = {P_B:.4f}")
print(f"P(A ∩ B) = {P_AB:.4f}")
print(f"P(A) · P(B) = {P_A * P_B:.4f}")
```

# 10. Probabilidad geométrica.

Esta es una extensión de la definición de probabilidad clásica, en donde ahora la probabilidad de un evento se calcula ya no a través de su cardinalidad, sino mediante la determinación de su **área, volumen o alguna característica geométrica equivalente**, según el problema que se trate.

Para el caso de áreas, se tiene la siguiente definición:

### Definición 1.3

Si un experimento aleatorio tiene como espacio muestral

$$
\Omega \subseteq \mathbb{R}^{2}
$$

cuya área está bien definida y es finita, entonces se define la **probabilidad geométrica** de un evento

$$
A \subseteq \Omega
$$

como

$$
P(A)=\frac{\text{Área de }A}{\text{Área de }\Omega},
$$

cuando el concepto de área del subconjunto \(A\) está bien definido.

Para poder aplicar la fórmula anterior, es necesario suponer que el espacio muestral es equiprobable en el sentido de que la probabilidad de observar la ocurrencia de un evento A depende únicamente de su área y no del conjunto
mismo.

### Ejemplo 1: Problema del encuentro de los dos amigos

Dos amigos acuerdan encontrarse en cierto lugar, pero olvidan la hora exacta de la cita. Únicamente recuerdan que esta ocurrirá entre las 12:00 y las 13:00 horas.

Cada persona decide llegar en un instante aleatorio dentro de ese intervalo y esperar únicamente **10 minutos**.

El objetivo es determinar la probabilidad de que ambos amigos logren encontrarse.

---

Sean:

$x$: el tiempo de llegada del primer amigo

y

$y$: el tiempo de llegada del segundo amigo, medidos en minutos después de las 12:00 horas

Como ambos pueden llegar en cualquier momento entre 0 y 60 minutos, el espacio muestral está formado por todos los puntos:

$$
\Omega=\{(x,y):0\leq x\leq60,\ 0\leq y\leq60\}.
$$

Geométricamente, este espacio corresponde a un cuadrado de lado 60 minutos, cuya área es:

$$
Área(\Omega)=60^2.
$$

Los amigos se encuentran cuando la diferencia entre sus tiempos de llegada es menor o igual a 10 minutos:

$$
|x-y|\leq10.
$$

La región donde ocurre este evento corresponde a todo el cuadrado excepto dos triángulos donde la diferencia entre tiempos supera los 10 minutos.

Cada uno de estos triángulos tiene lado:

$$
60-10=50.
$$

Por lo tanto, el área favorable es:

$$
Área(A)=60^2-50^2.
$$

Aplicando la definición de probabilidad geométrica:

$$
P(A)=\frac{Área(A)}{Área(\Omega)}
$$

obtenemos:

$$
P(A)=\frac{60^2-50^2}{60^2}
$$

$$
P(A)=\frac{3600-2500}{3600}
$$

$$
P(A)=\frac{1100}{3600}
$$

$$
P(A)=\frac{11}{36}\approx0.3056.
$$

Por lo tanto, la probabilidad de que los amigos se encuentren es aproximadamente $30.56\%$

```python
# ============================================
# Ejemplo: Problema de los dos amigos
# Probabilidad geométrica mediante simulación
# ============================================

import numpy as np

# Número de simulaciones
N = 1000000

# Llegadas aleatorias entre 0 y 60 minutos
amigo_1 = np.random.uniform(0, 60, N)
amigo_2 = np.random.uniform(0, 60, N)

# Evento: los amigos se encuentran
encuentro = np.abs(amigo_1 - amigo_2) <= 10

# Probabilidad estimada
probabilidad_simulada = np.mean(encuentro)

# Probabilidad teórica
probabilidad_teorica = 11/36

print("Probabilidad simulada:", probabilidad_simulada)
print("Probabilidad teórica:", probabilidad_teorica)
```

### Ejemplo 1: Generalizado

Tenemos el mismo problema donde dos personas acuerdan verse en un lugar durante la hora entre las 12:00 y la 13:00. Cada una llega en un instante aleatorio uniforme dentro de esa hora, y se encuentran solo si ninguna espera más de $\Delta$ (una fracción de hora) a la otra. Sean $X, Y \sim \text{Uniforme}(0, 1)$ los tiempos de llegada (en fracción de hora) de cada persona, independientes entre sí. Se encuentran si

$$|X - Y| < \Delta.$$

Este evento corresponde a una banda diagonal dentro del cuadrado unitario $[0,1]\times[0,1]$. Calculando el área de esa banda se obtiene la probabilidad teórica

$$P(|X-Y| < \Delta) = 1 - (1 - \Delta)^2.$$

Se compara este resultado con una simulación Monte Carlo.

```python
delta = 0.25  # ventana de encuentro: 15 minutos de una hora

p_teorica = 1 - (1 - delta) ** 2

rng = np.random.default_rng(3)
n_sim = 100_000
X = rng.uniform(0, 1, n_sim)
Y = rng.uniform(0, 1, n_sim)
se_encuentran = np.abs(X - Y) < delta
p_simulada = se_encuentran.mean()

print("Delta (ventana de encuentro):", delta)
print("Probabilidad teorica (area de la banda):", round(p_teorica, 4))
print("Probabilidad simulada:", round(p_simulada, 4))
```

```python
n_muestra_grafica = 3000
idx = rng.choice(n_sim, n_muestra_grafica, replace=False)

plt.figure(figsize=(6, 6))
plt.scatter(X[idx][se_encuentran[idx]], Y[idx][se_encuentran[idx]], s=4, color="tab:green", label="Se encuentran")
plt.scatter(X[idx][~se_encuentran[idx]], Y[idx][~se_encuentran[idx]], s=4, color="tab:gray", label="No se encuentran")
t = np.linspace(0, 1, 200)
plt.plot(t, np.clip(t + delta, 0, 1), color="black", linewidth=1)
plt.plot(t, np.clip(t - delta, 0, 1), color="black", linewidth=1)
plt.gca().set_aspect("equal")
plt.xlabel("Hora de llegada de la persona 1 (X)")
plt.ylabel("Hora de llegada de la persona 2 (Y)")
plt.title("Problema del encuentro: region |X - Y| < delta")
plt.legend(markerscale=3)
plt.show()
```

### Ejemplo 2.

Estimar $\pi$ generando puntos aleatorios uniformes en el cuadrado $[-1, 1] \times [-1, 1]$ y calculando la proporción que cae dentro del círculo unitario inscrito. Como el área del círculo es $\pi r^2 = \pi$ y la del cuadrado es 4, la proporción esperada de puntos dentro del círculo es $\pi / 4$.

```python
rng = np.random.default_rng(777)
n_puntos = 20_000
x = rng.uniform(-1, 1, n_puntos)
y = rng.uniform(-1, 1, n_puntos)
dentro_circulo = x**2 + y**2 <= 1

p_dentro = dentro_circulo.mean()
pi_estimado = p_dentro * 4

print("Proporcion de puntos dentro del circulo:", round(p_dentro, 4))
print("Estimacion de pi (4 * proporcion):", round(pi_estimado, 4))
print("Valor real de pi:", round(np.pi, 4))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 6))
plt.scatter(x[dentro_circulo], y[dentro_circulo], s=2, color="tab:blue", label="Dentro del circulo")
plt.scatter(x[~dentro_circulo], y[~dentro_circulo], s=2, color="tab:orange", label="Fuera del circulo")
plt.gca().set_aspect("equal")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Enfoque geometrico: estimacion de pi por Monte Carlo")
plt.legend(markerscale=5)
plt.show()
```

### Ejemplo 3: Problema del volado

Se lanza una moneda justa 5 veces. Bajo el enfoque clásico, el número de caras $k$ obtenido sigue la fórmula combinatoria

$$P(k \text{ caras}) = \binom{5}{k} (0.5)^k (0.5)^{5-k}, \qquad k = 0, 1, \dots, 5,$$

que es un caso particular de la distribución binomial. Se compara esta probabilidad teórica contra la frecuencia obtenida al simular muchas repeticiones de 5 volados.

```python
from math import comb

n_volados = 5
p_moneda = 0.5
k_valores = np.arange(0, n_volados + 1)
p_teorica_k = np.array([comb(n_volados, k) * p_moneda**k * (1 - p_moneda) ** (n_volados - k) for k in k_valores])

rng = np.random.default_rng(11)
n_repeticiones = 50_000
muestras = rng.integers(0, 2, size=(n_repeticiones, n_volados)).sum(axis=1)
frecuencia_simulada = np.array([(muestras == k).mean() for k in k_valores])

print("Numero de caras k:        ", k_valores)
print("Probabilidad teorica:     ", np.round(p_teorica_k, 4))
print("Frecuencia simulada:      ", np.round(frecuencia_simulada, 4))
```

```python
ancho = 0.35
fig, ax = plt.subplots()
ax.bar(k_valores - ancho / 2, p_teorica_k, ancho, label="Teorica")
ax.bar(k_valores + ancho / 2, frecuencia_simulada, ancho, label="Simulada")
ax.set_xticks(k_valores)
ax.set_xlabel("Numero de caras en 5 volados")
ax.set_ylabel("Probabilidad")
ax.set_title("Ejemplo del volado: numero de caras en 5 lanzamientos")
ax.legend()
plt.show()
```

### Ejemplo 4: Cuadrado unitario

Se escogen dos números $x$ y $y$ al azar dentro del intervalo unitario $[0,1]$. ¿Cúal es la probabilidad de que la suma de estos números sea mayor a uno y que, al mismo tiempo, la suma de sus cuadrados sea menor a uno?

**Solución**

La región favorable corresponde a la parte del cuarto de círculo que se encuentra por encima de la recta:

$$
x+y=1.
$$

El área del cuarto de círculo de radio 1 es:

$$
A_1=\frac{\pi}{4}.
$$

La región que queda fuera del evento corresponde a un triángulo rectángulo cuyos catetos tienen longitud 1, por lo que su área es:

$$
A_2=\frac{1}{2}.
$$

Por lo tanto, el área del evento \(A\) es:

$$
Área(A)=\frac{\pi}{4}-\frac{1}{2}.
$$

Como el área del espacio muestral es:

$$
Área(\Omega)=1,
$$

la probabilidad geométrica es:

$$
P(A)=\frac{Área(A)}{Área(\Omega)}
$$

$$
P(A)=\frac{\frac{\pi}{4}-\frac{1}{2}}{1}
$$

$$
\boxed{
P(A)=\frac{\pi}{4}-\frac{1}{2}
}
$$

Finalmente,

$$
P(A)\approx0.2854.
$$

Por lo tanto, la probabilidad de que los números seleccionados cumplan simultáneamente ambas condiciones es aproximadamente del $28.54\%$.

```python
# ============================================
# Probabilidad geométrica:
# x+y>1 y x^2+y^2<1
# ============================================

import numpy as np
import matplotlib.pyplot as plt

# Número de simulaciones
N = 200000

# Generación de puntos aleatorios dentro del cuadrado unitario
x = np.random.uniform(0, 1, N)
y = np.random.uniform(0, 1, N)

# Definición del evento
evento = (x + y > 1) & (x**2 + y**2 < 1)

# Probabilidad aproximada
probabilidad = np.mean(evento)

print("Probabilidad estimada:", probabilidad)
```

# 11. Probabilidad frecuentista

Suponga que se realizan $n$ repeticiones de un cierto experimento aleatorio y que se registra el número de veces que ocurre un determinado evento $A$. Esta información puede ser usada de la siguiente forma para definir la probabilidad de $A$:

Sea $n_A$ el número de ocurrencias de un evento $A$ en $n$ realizaciones de un experimento aleatorio. La **probabilidad frecuentista** del evento $A$ se define como el límite:

$$
P(A)=\lim_{n\rightarrow\infty}\frac{n_A}{n}.
$$

En este caso, debemos hacer notar que no es humanamente posible llevar a cabo una infinidad de veces el experimento aleatorio y tampoco podemos garantizar, por ahora, la existencia de tal límite.

Por lo tanto, mediante la definición anterior, no es posible encontrar de manera exacta la probabilidad de un evento cualquiera. Sin embargo, esta definición permite obtener una **aproximación empírica** del valor de $P(A)$, es decir:

$$
P(A)\approx \frac{n_A}{n}.
$$

La probabilidad frecuentista establece que, al aumentar el número de repeticiones del experimento, la frecuencia relativa de ocurrencia de un evento se aproxima a su verdadera probabilidad.

### Ejemplo 1: Lanzamientos de una moneda

Considérese el experimento aleatorio de lanzar una moneda equilibrada.

El evento de interés es:

$$
A=\{\text{obtener cara}\}.
$$

Teóricamente, la probabilidad de obtener cara es:

$$
P(A)=\frac{1}{2}=0.5.
$$

Desde el enfoque frecuentista, esta probabilidad puede aproximarse realizando un gran número de lanzamientos y calculando la frecuencia relativa de caras obtenidas:

$$
P(A)\approx\frac{n_A}{n},
$$

donde:

- $n_A$ representa el número de veces que aparece cara.
- $n$ representa el número total de lanzamientos realizados.

Para este ejemplo simulan $40,000$ lanzamientos de una moneda con ayuda de python. Después de cada lanzamiento se calcula la frecuencia acumulada de caras y se observa cómo esta proporción se acerca al valor teórico de $0.5$.

```python
rng = np.random.default_rng(677)
n_max = 40_000
volados = rng.integers(0, 2, size=n_max)  # 1 = cara, 0 = cruz

tamanos = np.arange(1, n_max + 1)
frecuencia_relativa = np.cumsum(volados) / tamanos

print("Frecuencia relativa de caras tras 10 volados:", round(frecuencia_relativa[9], 4))
print("Frecuencia relativa de caras tras 1000 volados:", round(frecuencia_relativa[999], 4))
print(f"Frecuencia relativa de caras tras {n_max} volados:", round(frecuencia_relativa[-1], 4))
```
